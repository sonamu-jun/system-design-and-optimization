## The Lagrange Multiplier Method

**Derive the multiplier equations and solve an exactly calibrated clock problem.**

Keep the clock observations and squared-error objective; add exact calibration, with no correction bound.

<style>

</style>

### 1 · The constrained clock

The fixed data vectors are $\boldsymbol{d}=(-2,-1,-1,0)$ min and $\boldsymbol{t}=(0,1,2,3)$ h. Their scalar components $d_i$ and $t_i$ give the error and time of reading $i$, for $i=1,\ldots,4$. Choose initial correction $q$ (min) and rate correction $r$ (min/h). The remaining errors and squared-error objective are

$$
y_i=d_i+q+rt_i,\qquad \phi(q,r)=\sum_{i=1}^4 y_i^2.
$$

Equations below use numerical values in these stated units.

Exact calibration at one hour requires $q+(1\,\mathrm h)r=1\,\mathrm{min}$, written $h=q+r-1=0$. No correction bound is imposed.

Feasibility comes first: the unrestricted minimum $(1.9,-0.6)$ has score $0.2$ min² but violates calibration by $0.3$ min. Along the feasible line $r=1-q$,

$$
\phi(q,1-q)=0.5+6(q-1.5)^2.
$$

Thus the unique feasible minimum is $(1.5,-0.5)$ with score $0.5$ min². The line misses the unrestricted minimum.

### 2 · Follow both settings along calibration

Changing $q$ while keeping calibration requires $r(q)=1-q$, so $dr/dq=-1$. For example, increasing $q$ by $0.1$ min requires decreasing $r$ by $0.1$ min/h. These paired changes follow the feasible line.

There are two different slopes:

| Derivative | What changes |
|:---|:---|
| $\partial\phi/\partial q$ | Change $q$ while holding $r$ fixed. |
| $d\phi(q,r(q))/dq$ | Change $q$ and also change $r$ to keep calibration. |

The second slope includes **both contributions**. The chain rule gives

$$
\frac{d}{dq}\phi(q,r(q))
=\frac{\partial\phi}{\partial q}\cdot 1
+\frac{\partial\phi}{\partial r}\frac{dr}{dq}.
$$

The first term is the direct change through $q$. The second is the change through $r$: its slope is multiplied by how fast $r$ changes. Substituting $dr/dq=-1$ gives

$$
\frac{d}{dq}\phi(q,1-q)
=\frac{\partial\phi}{\partial q}-\frac{\partial\phi}{\partial r}.
$$

Both directions along the line are allowed. At a smooth constrained minimum, this total slope is zero, so

$$
\frac{\partial\phi}{\partial q}
=\frac{\partial\phi}{\partial r}.
$$

The individual partial derivatives need not be zero. Their effects cancel along the allowed movement.

### 3 · Package that balance in the Lagrangian

Let $-\nu$ be the common value of those two partial derivatives. Then

$$
\frac{\partial\phi}{\partial q}+\nu=0,
\qquad
\frac{\partial\phi}{\partial r}+\nu=0.
$$

The same $\nu$ appears twice because the coefficients of $q$ and $r$ in $h=q+r-1$ are both $1$. The **Lagrangian** packages this slope balance and the calibration requirement:

$$
L(q,r,\nu)=\phi(q,r)+\nu(q+r-1).
$$

In each partial derivative, hold the other two variables fixed:

$$
\begin{aligned}
\frac{\partial L}{\partial q}
&=\frac{\partial\phi}{\partial q}+\nu\cdot1=0,\\
\frac{\partial L}{\partial r}
&=\frac{\partial\phi}{\partial r}+\nu\cdot1=0,\\
\frac{\partial L}{\partial\nu}
&=q+r-1=0.
\end{aligned}
$$

The first two equations express stationarity; the third enforces feasibility. The multiplier $\nu$ has no sign restriction and is solved with $q$ and $r$. It is not an analyst-chosen score weight. The Lagrangian is not minimized freely over all three variables.

### 4 · Calculate each partial derivative and solve

The four remaining errors are $y_1=q-2$, $y_2=q+r-1$, $y_3=q+2r-1$, and $y_4=q+3r$. Each contributes $y_i^2$ to $\phi$.

Differentiating a squared error gives $2y_i$ times the derivative of its inside expression. For $q$, those inside derivatives are all $1$. For $r$, they are $0,1,2,3$:

$$
\begin{aligned}
\frac{\partial\phi}{\partial q}
&=2(y_1+y_2+y_3+y_4)\\
&=8q+12r-8,\\[0.4em]
\frac{\partial\phi}{\partial r}
&=2(0y_1+1y_2+2y_3+3y_4)\\
&=12q+28r-6.
\end{aligned}
$$

Substitution into the three Lagrangian equations gives

$$
\begin{aligned}
8q+12r-8+\nu&=0,\\
12q+28r-6+\nu&=0,\\
q+r-1&=0.
\end{aligned}
$$

Subtracting the first equation from the second gives $4q+16r+2=0$. With $q=1-r$, this becomes $6+12r=0$. Thus $r=-0.5$, $q=1.5$, and $\nu=2$.

At this point, $\partial\phi/\partial q=-2$ and $\partial\phi/\partial r=-2$. Along calibration, the two contributions cancel: $(-2)\cdot1+(-2)\cdot(-1)=0$. In the Lagrangian, adding $\nu=2$ cancels each partial derivative separately, as shown in the two rows.

Stationarity is necessary here. The identity $\phi(q,1-q)=0.5+6(q-1.5)^2$ separately proves the unique minimum.

### 5 · Interactive comparison

The $q$ slider moves along calibration, with $r=1-q$. The $\nu$ slider changes only the Lagrangian slope balance. At $q=1.5$, $\nu=2$, both stationarity residuals vanish. At other calibrated points, a multiplier cannot cancel both slopes. The figure keeps the objective and feasible line fixed.

```python
import sys
import numpy as np
import matplotlib

if sys.platform != "emscripten":
    try:
        matplotlib.use("widget", force=True)
    except (RuntimeError, ValueError):
        from matplotlib.backends import backend_registry
        backend_registry._clear()
        matplotlib.use("widget", force=True)
import matplotlib.pyplot as plt
from matplotlib.widgets import Slider

BLUE, TEAL, ORANGE, PURPLE, GRAY = "#2878b5", "#168578", "#e78b24", "#8854a5", "#707070"
plt.rcParams.update({"font.size": 11, "axes.titlesize": 11})

# Fixed observations and the physical response.
observed_errors = np.array([-2., -1., -1., 0.])  # min
observation_times = np.arange(4.)  # h

def evaluate_clock(decision):
    correction, rate = decision
    errors = observed_errors + correction + rate * observation_times
    gradient = 2 * np.array([errors.sum(), observation_times @ errors])
    return {"decision": np.array(decision), "errors": errors,
            "objective": float(errors @ errors), "gradient": gradient}
```

```python
figure, axis = plt.subplots(figsize=(5.6, 4.8))
figure.subplots_adjust(left=0.15, right=0.96, top=0.90, bottom=0.43)
corrections = np.linspace(0.5, 2.5, 200)
axis.plot(corrections, 0.5 + 6*(corrections-1.5)**2, color=BLUE)
axis.plot(1.5, 0.5, "o", color=TEAL)
candidate, = axis.plot([], [], "o", color=ORANGE)
axis.set(xlabel="Correction q (min)", ylabel="Squared error (min²)",
         title="Feasible minimum and slope balance")
axis.grid(alpha=0.25)
status = figure.text(0.15, 0.27, "", va="top")
q_slider = Slider(figure.add_axes([0.25, 0.12, 0.58, 0.035]), "q (min)",
                  0.5, 2.5, valinit=1.0, valstep=0.05, color=BLUE)
nu_slider = Slider(figure.add_axes([0.25, 0.05, 0.58, 0.035]), "nu",
                   -6, 10, valinit=2, valstep=0.1, color=ORANGE)

def update(_):
    q, nu = q_slider.val, nu_slider.val
    result = evaluate_clock((q, 1-q))
    residual = result["gradient"] + nu*np.ones(2)
    tangent_slope = result["gradient"] @ np.array([1., -1.])
    candidate.set_data([q], [result["objective"]])
    status.set_text(f"r = {1-q:.2f} min/h; score = {result['objective']:.3f} min²\n"
                    f"Feasible slope = {tangent_slope:.2f}\n"
                    f"L slopes (q, r) = ({residual[0]:.2f}, {residual[1]:.2f})")
    figure.canvas.draw_idle()

figure._sliders = [q_slider, nu_slider]
for slider in figure._sliders:
    slider.on_changed(update)
update(None)
plt.show()
```
