## The Lagrange Multiplier Method

**At a constrained minimum, no small allowed movement lowers the score to first order.**

Keep the clock observations and squared-error objective. Exact calibration restricts the two settings to a line; there is no correction bound. Small changes explain why a multiplier appears.


<style>
table { margin-left: 0 !important; margin-right: auto !important; }
th, td { text-align: left !important; }
</style>

### 1 · The clock decision and its constraint

Choose initial correction $q$ (min) and rate correction $r$ (min/h). The four recorded errors are $d_i=-2,-1,-1,0$ min at times $t_i=0,1,2,3$ h, for $i=1,\ldots,4$. The remaining errors and score are

$$
y_i=d_i+q+rt_i,\qquad \phi(q,r)=\sum_{i=1}^4 y_i^2.
$$

Using numerical values in these units, exact calibration at one hour requires $h(q,r)=q+r-1=0$. Only settings on the line $r=1-q$ are allowed. The unrestricted minimum misses that line; the best calibrated setting lies on it.

The same problem can be written generally as

$$
\min_x f(x)\qquad\text{subject to }h(x)=0.
$$

Here $x$ is the decision vector, $f$ the objective, and $h$ the equality residual. For the clock, $x=(q,r)^{\mathsf T}$ and $f(x)=\phi(q,r)$.

### 2 · Write the first-order change in the score

For three components, write the decision and a small change as

$$
x=\begin{bmatrix}x_1\\x_2\\x_3\end{bmatrix},\qquad
\delta x=\begin{bmatrix}\delta x_1\\\delta x_2\\\delta x_3\end{bmatrix}.
$$

Each component contributes its partial derivative times its change. Adding gives the **first-order change**, denoted by $\delta f$:

$$
\begin{aligned}
\delta f
&=\frac{\partial f}{\partial x_1}\delta x_1
 +\frac{\partial f}{\partial x_2}\delta x_2
 +\frac{\partial f}{\partial x_3}\delta x_3\\
&=\nabla f(x)^{\mathsf T}\delta x.
\end{aligned}
$$

The gradient $\nabla f$ is the column of partial derivatives. The dot product adds the three contributions. The actual change $f(x+\delta x)-f(x)$ is approximately $\delta f$ for a small movement; it need not equal it exactly. The same calculation applies to the clock's two components.

### 3 · Keep the movement along the constraint

Start at an allowed point, so $h(x)=0$. A small change gives

$$
h(x+\delta x)\approx h(x)+\nabla h(x)^{\mathsf T}\delta x.
$$

To keep the constraint satisfied to first order, the extra term must be zero:

$$
\nabla h(x)^{\mathsf T}\delta x=0.
$$

Thus an allowed **tangent direction** is perpendicular to $\nabla h$. For a curved constraint, this describes the tangent, not an exact finite movement along the curve.

For the clock, $h=q+r-1$ is linear, so the condition is simply $\delta q+\delta r=0$. Increasing $q$ by $0.1$ min and decreasing $r$ by $0.1$ min/h keeps calibration exactly.

### 4 · At the minimum, both gradients are normal to the constraint

At a smooth constrained minimum $x^\star$, both signs of each tangent direction are allowed. If one sign decreased the score to first order, the point would not be a minimum. Therefore

$$
\nabla f(x^\star)^{\mathsf T}\delta x=0
\quad\text{for every allowed tangent direction }\delta x.
$$

Both $\nabla f$ and $\nabla h$ are perpendicular to all those directions. With one equality and $\nabla h(x^\star)\ne0$, this means $\nabla f$ is a multiple of $\nabla h$. Name the multiple $-\nu$:

$$
\nabla f(x^\star)=-\nu\nabla h(x^\star)
\quad\Longleftrightarrow\quad
\nabla f(x^\star)+\nu\nabla h(x^\star)=0.
$$

The nonzero gradients are parallel, possibly pointing in opposite directions. The statement uses **every** tangent direction, not just one chosen movement.

For the clock optimum $(1.5,-0.5)$, the gradients are $\nabla\phi=(-2,-2)^{\mathsf T}$ and $\nabla h=(1,1)^{\mathsf T}$. Both are perpendicular to the calibration line; $\nu=2$ balances them. The arrows show directions in the numerical $(q,r)$ coordinates.

### 5 · Put the two conditions in one function

The **Lagrangian** is defined by

$$
L(x,\nu)=f(x)+\nu h(x).
$$

Its derivatives reproduce the slope balance and the constraint:

$$
\nabla_{(x,\nu)}L
=\begin{bmatrix}
\nabla f(x)+\nu\nabla h(x)\\
h(x)
\end{bmatrix}
=0.
$$

The upper block differentiates with respect to the decision components; the last entry differentiates with respect to $\nu$. Solve these equations together for $x$ and $\nu$.

**The derivatives are zero; the value of $L$ need not be zero.** On an allowed point, $h=0$, so $L=f$. The multiplier has no sign restriction and is solved, not chosen as a score weight. These are stationary equations, not a free minimization over $(x,\nu)$.

### 6 · Solve the clock equations

For $L=\phi+\nu(q+r-1)$, the two objective partial derivatives are

$$
\frac{\partial\phi}{\partial q}=8q+12r-8,\qquad
\frac{\partial\phi}{\partial r}=12q+28r-6.
$$

Thus the three equations are

$$
\begin{aligned}
8q+12r-8+\nu&=0,\\
12q+28r-6+\nu&=0,\\
q+r-1&=0.
\end{aligned}
$$

Subtract the first row from the second: $4q+16r+2=0$. Using $q=1-r$ gives $6+12r=0$. Hence $r=-0.5$, $q=1.5$, and $\nu=2$.

Each objective partial derivative is $-2$, canceled by the constraint contribution $+2$. The value is $L=\phi=0.5$ min², not zero.

Stationarity alone need not prove a minimum in general. Here a direct check along calibration does:

$$
\phi(q,1-q)=0.5+6(q-1.5)^2.
$$

The squared term is nonnegative, so $(1.5,-0.5)$ is the unique allowed minimum.

### 7 · Interactive comparison

The $q$ slider moves along calibration, with $r=1-q$. The $\nu$ slider changes only the Lagrangian slope balance. At $q=1.5$, $\nu=2$, both stationarity residuals vanish. At other calibrated points, a multiplier cannot cancel both slopes. The figure keeps the objective and feasible line fixed.


```python
import sys
import numpy as np
import matplotlib

if sys.platform != "emscripten":
    try:
        matplotlib.use("widget", force=True)
    except (RuntimeError, ValueError):
        from matplotlib.backends import backend_registry
        backend_registry._clear()
        matplotlib.use("widget", force=True)
import matplotlib.pyplot as plt
from matplotlib.widgets import Slider

BLUE, TEAL, ORANGE, PURPLE, GRAY = "#2878b5", "#168578", "#e78b24", "#8854a5", "#707070"
plt.rcParams.update({"font.size": 11, "axes.titlesize": 11})

# Fixed observations and the physical response.
observed_errors = np.array([-2., -1., -1., 0.])  # min
observation_times = np.arange(4.)  # h

def evaluate_clock(decision):
    correction, rate = decision
    errors = observed_errors + correction + rate * observation_times
    gradient = 2 * np.array([errors.sum(), observation_times @ errors])
    return {"decision": np.array(decision), "errors": errors,
            "objective": float(errors @ errors), "gradient": gradient}
```

```python
figure, axis = plt.subplots(figsize=(5.6, 4.8))
figure.subplots_adjust(left=0.15, right=0.96, top=0.90, bottom=0.43)
corrections = np.linspace(0.5, 2.5, 200)
axis.plot(corrections, 0.5 + 6*(corrections-1.5)**2, color=BLUE)
axis.plot(1.5, 0.5, "o", color=TEAL)
candidate, = axis.plot([], [], "o", color=ORANGE)
axis.set(xlabel="Correction q (min)", ylabel="Squared error (min²)",
         title="Feasible minimum and slope balance")
axis.grid(alpha=0.25)
status = figure.text(0.15, 0.27, "", va="top")
q_slider = Slider(figure.add_axes([0.25, 0.12, 0.58, 0.035]), "q (min)",
                  0.5, 2.5, valinit=1.0, valstep=0.05, color=BLUE)
nu_slider = Slider(figure.add_axes([0.25, 0.05, 0.58, 0.035]), "nu",
                   -6, 10, valinit=2, valstep=0.1, color=ORANGE)

def update(_):
    q, nu = q_slider.val, nu_slider.val
    result = evaluate_clock((q, 1-q))
    residual = result["gradient"] + nu*np.ones(2)
    tangent_slope = result["gradient"] @ np.array([1., -1.])
    candidate.set_data([q], [result["objective"]])
    status.set_text(f"r = {1-q:.2f} min/h; score = {result['objective']:.3f} min²\n"
                    f"Feasible slope = {tangent_slope:.2f}\n"
                    f"L slopes (q, r) = ({residual[0]:.2f}, {residual[1]:.2f})")
    figure.canvas.draw_idle()

figure._sliders = [q_slider, nu_slider]
for slider in figure._sliders:
    slider.on_changed(update)
update(None)
plt.show()
```
