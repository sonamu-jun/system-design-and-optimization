## 휴리스틱 개념 (Heuristic Concepts)

<strong>휴리스틱은 최적성을 보장하지는 않지만, 양질의 실행 가능한 해를 효율적으로 탐색하는 방법이다.</strong>

### 1 · 휴리스틱의 개념

- <strong>탐색 원리:</strong> 문제별 규칙으로 후보해를 생성·선택·개선한다.
- <strong>계산 효율:</strong> 탐색 범위를 제한하여 계산량을 줄인다. 전역 최적성은 보장하지 않는다.
- <strong>해의 평가:</strong> 제약조건을 만족하는 실행 가능한 해(feasible solution)를 목적함수 값으로 비교한다.

- <strong>예시: 배송 경로 최적화</strong>
  - 의사결정변수: 다섯 도시의 방문 순서.
  - 제약조건: 각 도시를 한 번씩 방문한 뒤 출발점으로 복귀.
  - 목적함수: 복귀 구간을 포함한 총거리.
  - 아래 세 후보는 동일한 도시 배치와 이동 거리를 사용한다. 원은 도시, 숫자는 구간 거리이며 선 길이는 실제 거리와 무관하다.

  | 후보 | 경로 | 거리 | 평가 |
  |:---|:---|---:|:---|
  | A | 1→2→0→3→4→1 | 34 | 실행 가능 |
  | B | 1→0→4→3→2→1 | 22 | A보다 우수한 실행 가능해 |
  | C | 1→0→4→3→2 | 14 | 복귀 누락으로 실행 불가능 |

  - <strong>후보 A:</strong> 방문 거리 28 + 주황색 복귀 거리 6 = 총거리 34.

  <div style="text-align: left; margin: 0.65rem 0 1.5rem 2.5rem;"><img src="https://raw.githubusercontent.com/sonamu-jun/system-design-and-optimization/main/05-1_heuristic_optimization_greedy_and_local_search/assets/longer_route.svg" alt="후보 A: 1→2→0→3→4→1. 구간 거리 8, 9, 10, 1, 6의 합은 34이며 모든 방문과 복귀 조건을 만족한다." width="440" style="display: block; width: 27.5em; max-width: 100%; height: auto; margin: 0;"></div>

  - <strong>후보 B:</strong> 방문 거리 14 + 주황색 복귀 거리 8 = 총거리 22. A와 같은 조건을 만족하면서 거리가 12만큼 짧다.

  <div style="text-align: left; margin: 0.65rem 0 1.5rem 2.5rem;"><img src="https://raw.githubusercontent.com/sonamu-jun/system-design-and-optimization/main/05-1_heuristic_optimization_greedy_and_local_search/assets/feasible_return.svg" alt="후보 B: A와 동일한 도시 배치에서 1→0→4→3→2→1로 방문한다. 구간 거리 2, 7, 1, 4, 8의 합은 22이며 A보다 12만큼 짧다." width="440" style="display: block; width: 27.5em; max-width: 100%; height: auto; margin: 0;"></div>

  - <strong>후보 C:</strong> 거리는 14지만 회색 점선의 복귀 구간이 누락되었다. 제약조건을 위반하므로 더 우수한 해로 평가할 수 없다.

  <div style="text-align: left; margin: 0.65rem 0 1.5rem 2.5rem;"><img src="https://raw.githubusercontent.com/sonamu-jun/system-design-and-optimization/main/05-1_heuristic_optimization_greedy_and_local_search/assets/incomplete_route.svg" alt="후보 C: 같은 도시 배치에서 1→0→4→3→2까지만 이동하여 거리는 14이다. 회색 점선은 누락된 2→1 복귀 구간이며 이 후보는 실행 불가능하다." width="440" style="display: block; width: 27.5em; max-width: 100%; height: auto; margin: 0;"></div>

  - <strong>비교 원칙:</strong> 실행 가능성을 먼저 확인한 뒤 목적함수 값을 비교한다. B가 A보다 우수하다는 사실만으로 B의 전역 최적성이 보장되지는 않는다.

### 2 · 수학적 최적화와 휴리스틱의 필요성

- <strong>문제 정식화:</strong> 의사결정변수·목적함수·제약조건으로 최적화 문제를 표현한다.
- <strong>해법의 적용 조건:</strong> 볼록성이나 미분 가능성은 해법 선택과 최적성 판단에 중요하다. 수식으로 표현할 수 있어도 효율적인 해 계산이 보장되지는 않는다.

| 문제의 특성 | 해를 구할 때의 어려움 |
|:---|:---|
| 비볼록성(nonconvexity) | 여러 지역 최적해가 존재할 수 있어 전역 최적해 탐색이 어려움 |
| 미분 정보의 이용 어려움 | 목적함수가 미분 불가능하거나 평가 결과에 잡음이 있으면 도함수의 계산·추정이 어려울 수 있음 |
| 이산적인 의사결정 | 조합 수가 증가하면 모든 후보의 목적함수 값을 계산·비교하는 비용이 커짐 |

- <strong>휴리스틱의 역할:</strong> 문제별 규칙으로 후보해를 생성·평가·개선한다. 미분 정보 없이 목적함수 값을 비교하는 방식도 사용할 수 있다.
- <strong>선택 기준:</strong> 최적성 보장보다 제한된 계산 시간 안에서 양질의 실행 가능해를 얻는 것이 중요할 때 활용한다.
- 비볼록·미분 불가능 문제에도 전용 해법이 있다. 휴리스틱은 이러한 문제를 푸는 여러 접근 중 하나이며, 항상 더 효율적인 것은 아니다.

- <strong>예시: 이산적 의사결정의 전체 후보 비교 — 배송 경로</strong>
  - 도시 방문 순서는 이산적인 의사결정이다. 아래는 가능한 순서를 모두 평가하는 <strong>완전 탐색(exhaustive search)</strong>의 계산 부담을 보여 준다.
  - 각 후보의 총거리를 계산한 뒤, 모든 후보 중 최솟값을 선택한다.
  - 출발 도시를 고정해도 도시 수가 늘면 비교할 방문 순서는 빠르게 증가한다.

  - 평가 시간은 경로 하나당 <strong>1밀리초(0.001초)</strong>로 가정한다. 모든 후보를 순차 평가하며, 역방향 순서도 별도 후보로 센다.

  | 도시 수 | 비교할 방문 순서 | 전체 평가 시간 |
  |:---|:---|:---|
  | 5개 | $4!=24$개 | 0.024초 |
  | 10개 | $9!=362{,}880$개 | 약 6분 |
  | 15개 | $14!\approx8.72\times10^{10}$개 | 약 2.8년 |
  | 20개 | $19!\approx1.22\times10^{17}$개 | 약 385만 년 |

  - <strong>개별 평가가 빨라도 조합 수가 급증하면 전체 비교는 현실적으로 어려워진다.</strong>
  - 1밀리초는 설명을 위한 고정 가정이다. 표는 완전 탐색의 예상 시간이며, 모든 최적화 해법에 필요한 시간을 뜻하지 않는다.
  - 휴리스틱은 일부 후보만 평가하도록 탐색을 제한한다. 대신 전역 최적성은 보장하지 않는다.

### 3 · 휴리스틱과 메타휴리스틱

- <strong>문제 특화 휴리스틱(problem-specific heuristic):</strong> 특정 문제의 구조와 제약에 맞춘 후보해 생성·선택·개선 규칙.
- <strong>메타휴리스틱(metaheuristic):</strong> 특정 문제에 종속되지 않는 범용 탐색 전략. 탐색의 진행·재시작·후보 수용 등을 제어한다.

| 구분 | 문제 특화 휴리스틱 | 메타휴리스틱 |
|:---|:---|:---|
| 설계 중심 | 해당 문제의 특성을 이용한 규칙 | 여러 문제에 재사용할 탐색 전략 |
| 적용 범위 | 해당 문제 또는 유사한 구조의 문제 | 다양한 최적화 문제 |
| 다른 문제에 적용 | 문제에 맞게 규칙 수정 | 전략은 유지하되 해 표현·후보 생성·평가 방식 설계 |

- <strong>범용성의 의미:</strong> 메타휴리스틱은 다양한 문제에 맞게 구현할 수 있다. 어떤 문제에도 수정 없이 적용되거나 좋은 성능이 보장된다는 뜻은 아니다.
- 메타휴리스틱도 넓은 의미의 휴리스틱에 포함된다. 문제 특화 규칙을 구성 요소로 활용할 수 있으며, 전역 최적성은 보장하지 않는다.

### 4 · 대표적인 휴리스틱 알고리듬

- 이 단원의 방법을 <strong>해 구성 → 해 개선</strong>의 역할에 따라 정리한다.
- <strong>해 구성:</strong> 부분해를 확장하여 실행 가능한 해를 완성한다.
  - <strong>탐욕 알고리듬(greedy algorithm):</strong> 매 단계에서 현재 가장 유리한 선택을 하는 일반적인 설계 방식이다.
    - <strong>최근접 이웃 알고리듬:</strong> 탐욕적 선택을 이용하여 현재 도시에서 가장 가까운 미방문 도시를 추가한다.
    - <strong>Cheapest insertion:</strong> 삽입 휴리스틱 중 비용 증가가 최소인 요소와 위치를 선택하는 방법이다.
- <strong>해 개선:</strong> 기존 해를 수정하여 목적함수 값을 개선한다.
  - <strong>지역 탐색(local search):</strong> 정해진 변경으로 생성한 이웃 해를 탐색한다.
    - <strong>2-opt 알고리듬:</strong> 경로의 두 연결을 교환하여 이웃을 생성한다. 여기서는 더 짧은 경로가 있을 때만 갱신한다.

- 위 구분은 이 단원에서 사용하는 관계이다. 모든 구성 방법이 탐욕적이거나 모든 지역 탐색이 개선되는 후보만 수용하는 것은 아니다.
