## 최근접 이웃 알고리듬 (Nearest Neighbor Algorithm)

<strong>최근접 이웃 알고리듬은 탐욕적 선택을 이용한 경로 구성 방법이다.</strong>

### 1 · 핵심 원리

- <strong>선택 규칙:</strong> 현재 도시에서 가장 가까운 미방문 도시를 선택한다.
- <strong>탐욕 알고리듬과의 관계:</strong> 탐욕 알고리듬은 매 단계에서 현재 가장 유리한 선택을 하는 일반적인 설계 방식이다. 최근접 이웃은 이를 거리 기준으로 적용한 구체적인 알고리듬이다.
- <strong>목적:</strong> 모든 방문 순서를 비교하지 않고 경로를 빠르게 구성한다.

### 2 · 적용 예시

- <strong>예시: 다섯 도시의 배송 경로</strong>
  - 도시 0–4를 한 번씩 방문하고 출발 도시 1로 복귀한다. 목적함수는 총거리이다.
  - 고정된 대칭 거리는 코드의 `distance`에 주어진다. `distance[i][j]`는 도시 i에서 j까지의 거리이다.
  - 도시 1에서 0·2·3·4까지의 거리는 2·8·5·6이다. 가장 가까운 도시 0을 선택한다.
  - 같은 규칙을 반복하면 $1\to0\to4\to3\to2\to1$, 총거리는 $2+7+1+4+8=22$이다.
  - 그림의 숫자는 거리이며, 선 길이는 실제 거리와 무관하다.

### 3 · 구현

- `unvisited`에서 가장 가까운 도시를 골라 경로에 추가한다. 동률이면 도시 번호가 작은 쪽을 선택한다.
- 모든 도시를 방문하면 출발 도시를 덧붙여 종료한다.


```python
distance = [
    [0, 2, 9, 10, 7],
    [2, 0, 8, 5, 6],
    [9, 8, 0, 4, 3],
    [10, 5, 4, 0, 1],
    [7, 6, 3, 1, 0],
]

def tour_cost(tour):
    total = 0
    for i in range(len(tour) - 1):
        a = tour[i]
        b = tour[i + 1]
        total += distance[a][b]
    return total

start = 1
tour = [start]
unvisited = list(range(len(distance)))
unvisited.remove(start)
while unvisited:
    current_city = tour[-1]
    next_city = unvisited[0]
    for city in unvisited:
        if distance[current_city][city] < distance[current_city][next_city]:
            next_city = city

    tour.append(next_city)
    unvisited.remove(next_city)
tour.append(start)

print("Route:", tour)
print("Distance:", tour_cost(tour))
```

- <strong>결과:</strong> `[1, 0, 4, 3, 2, 1]`, 총거리 22.
- <strong>해의 품질:</strong> 24개 방문 순서를 모두 평가한 최적값은 20이다. 최근접 이웃의 경로는 최적값보다 10% 길다.
- <strong>의미와 한계:</strong> 적은 탐색으로 최적값에 가까운 해를 얻은 사례이다. 다만 가까운 도시를 먼저 선택하면 이후 방문·복귀 비용이 커질 수 있어, 다른 문제에서도 같은 품질이나 최적성을 보장하지는 않는다.
