## Recall the Clock Problem

**Recall the decision and its score before using derivatives to find the best correction.**

Lectures 01-2 and 02-1 introduced the clock decision and its objective. Lecture 02-2 supplied derivatives and partial derivatives. We keep the same four observations and now use those tools to calculate a solution and check that it is a minimum.

### 1 · Start with one clock setting

The four recorded clock errors are $-2,-1,-1,0$ minutes. A negative error means the clock is slow.
We choose one correction $q$, in minutes, and add it to every reading.

Follow the same 1 min correction through all four observations. The dial is the choice we make; the remaining errors are its results. The scorecard adds their squares so we can compare this choice with another.

| Correction | Remaining errors (min) | Sum of squared errors (min²) |
|:---|:---|---:|
| $q=0$ | $-2,-1,-1,0$ | $4+1+1+0=6$ |
| $q=1$ | $-1,0,0,1$ | $1+0+0+1=2$ |

The correction is the **decision**. The sum of squares is the **objective** used to compare decisions.
The second choice is better among these two. To show that it is best among all real corrections, we need another argument; [Part 2](02_single_variable_analytical_optimization.ipynb) supplies it.

### 2 · Find the best correction

Every real correction is allowed; there are no constraints in this example. Comparing a few corrections does not prove which one is best over the real line.

An **analytical method** uses equations to find a solution. In [Part 2](02_single_variable_analytical_optimization.ipynb), the derivative gives a candidate, and a separate check proves it is a minimum.
