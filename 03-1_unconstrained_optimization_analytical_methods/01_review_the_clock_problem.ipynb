### 1 · Start with one clock setting

The four recorded clock errors are $-2,-1,-1,0$ minutes. A negative error means the clock is slow.
We choose one correction $q$, in minutes, and add it to every reading.

Follow the same 1 min correction through all four observations. The dial is the choice we make; the remaining errors are its results. The scorecard adds their squares so we can compare this choice with another.

| Correction | Remaining errors (min) | Sum of squared errors (min²) |
|:---|:---|---:|
| $q=0$ | $-2,-1,-1,0$ | $4+1+1+0=6$ |
| $q=1$ | $-1,0,0,1$ | $1+0+0+1=2$ |

The correction is the **decision**. The sum of squares is the **objective** used to compare decisions.
The second choice is better among these two. To show that it is best among all real corrections, we need another argument; [Part 2](02_single_variable_analytical_optimization.ipynb) supplies it.

### 2 · Compare a few corrections

Every real correction is allowed; there are no constraints in this example.
Calculate the remaining errors and add their squares. The function returns both the errors and their score.


```python
import numpy as np

# Fixed observations, in minutes
OBSERVED_ERRORS = np.array([-2.0, -1.0, -1.0, 0.0])


def evaluate_clock(correction):
    remaining_errors = OBSERVED_ERRORS + correction
    return {
        "decision": correction,
        "response": remaining_errors,
        "objective": float(np.sum(remaining_errors ** 2)),
    }
```

Compare three corrections. Each printed row uses the same four observations.


```python
for correction in (0.0, 1.0, 2.0):
    result = evaluate_clock(correction)
    print(f"q = {correction:.1f} min | errors = {result['response']} min"
          f" | squared-error score = {result['objective']:.1f} min²")
```

### 3 · Practice with a worked solution

**Question.** At $q=0.5$, what are the remaining errors and squared-error score? Does this comparison alone prove that $q=1$ is best among all real corrections?

**Solution.** The errors are $-1.5,-0.5,-0.5,0.5$ min. Their squares add to $2.25+0.25+0.25+0.25=3$ min².
The score is larger than 2 min² at $q=1$, but comparing a few choices does not prove a minimum over all real corrections.


```python
practice_result = evaluate_clock(0.5)
print("Remaining errors:", practice_result["response"], "min")
print("Squared-error score:", practice_result["objective"], "min²")
```
