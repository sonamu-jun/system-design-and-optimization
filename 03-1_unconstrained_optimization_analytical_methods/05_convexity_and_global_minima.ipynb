## Convexity and Global Minima

**Use the shape of the clock score to understand why a local minimum is also a global minimum.**

[Part 4](04_least_squares.ipynb) interpreted squared errors. We keep the same four clock observations and return to one unrestricted correction. We now name the structure that made the minimum easy to verify: convexity.

### 1 · Keep the decision and score fixed

The recorded errors are $d_i=-2,-1,-1,0$ min, for $i=1,\ldots,4$. Choose a correction $q\in\mathbb R$, in minutes. The remaining errors are $y_i=d_i+q$. Their squared-error score is

$$
\phi(q)=\sum_{i=1}^{4}(d_i+q)^2=4(q-1)^2+2.
$$

The score is measured in min². At $q=0$ and $q=2$, it is 6. At their midpoint $q=1$, it is 2, which is below the average endpoint score of 6. We will extend this comparison to any two corrections.

The definitions below collect the fixed observations, score calculation, and graph of the score and its chord.


```python
import numpy as np
import matplotlib.pyplot as plt

OBSERVED_ERRORS = np.array([-2.0, -1.0, -1.0, 0.0])  # min


def evaluate_clock(correction):
    errors = OBSERVED_ERRORS + correction
    return {"decision": correction, "response": errors,
            "objective": float(errors @ errors)}


def show_convex_score():
    corrections = np.linspace(-0.25, 2.25, 201)
    scores = [evaluate_clock(q)["objective"] for q in corrections]
    figure, axis = plt.subplots(figsize=(7.5, 4.5))
    axis.plot(corrections, scores, color="#2878b5", label="Clock score")
    axis.plot([0, 2], [6, 6], "--", color="#909090", label="Endpoint interpolation")
    axis.scatter([0, 2], [6, 6], color="#e78b24", label="Two candidate settings", zorder=3)
    axis.scatter([1], [2], color="#168578", marker="D", label="Verified minimum", zorder=4)
    axis.set(title="The convex score lies below its endpoint chord",
             xlabel="Correction q (min)", ylabel="Squared-error score (min²)")
    axis.grid(alpha=0.25)
    axis.legend(fontsize=9)
    figure.tight_layout()
    return figure
```

### 2 · Compare a mixture of two corrections

Choose any two corrections $a,b\in\mathbb R$. Here $a$ and $b$ name candidate corrections, not new clock parameters. Let $\theta\in[0,1]$ specify a point between them. A function is **convex** when

$$
\phi\bigl(\theta a+(1-\theta)b\bigr)
\le \theta\phi(a)+(1-\theta)\phi(b)
$$

for every such choice. The score at the intermediate correction is no higher than the corresponding straight-line interpolation of endpoint scores. The intermediate correction is a single real setting, not a random selection of endpoints.

For this clock score, subtraction gives

$$
\begin{aligned}
&\theta\phi(a)+(1-\theta)\phi(b)\\
&\quad-\phi\bigl(\theta a+(1-\theta)b\bigr)
=4\theta(1-\theta)(a-b)^2\ge0.
\end{aligned}
$$

For $a\ne b$ and $0<\theta<1$, the difference is positive. This is **strict convexity**.

The next calculation checks the midpoint example. The identity above, rather than this one calculation, establishes convexity for every pair.


```python
left, right, weight = 0.0, 2.0, 0.5
midpoint = weight * left + (1 - weight) * right
interpolated_score = (weight * evaluate_clock(left)["objective"]
                      + (1 - weight) * evaluate_clock(right)["objective"])
print("Score at intermediate correction:", evaluate_clock(midpoint)["objective"], "min²")
print("Interpolated endpoint score:", interpolated_score, "min²")
```

### 3 · Connect the shape to optimality

A **convex set** contains the line segment between every pair of its points. The real line is convex. So are an interval and the equality line $q+r=1$ used in the next lecture, where $r$ is the hourly rate correction expressed numerically in min/h.

A **convex optimization problem** minimizes a convex function over a convex feasible set. Every local minimum of such a problem is global: if a better feasible point existed, the segment toward it would give better points arbitrarily nearby. Strict convexity gives at most one minimizer; it does not by itself guarantee that a minimizer exists.

Here $\phi'(q)=8(q-1)$ is zero at $q=1$. The identity $\phi(q)=4(q-1)^2+2$ proves that this is the unique global minimum.

| Description | Meaning for this clock problem |
|:---|:---|
| Unconstrained | Every real correction is allowed |
| Nonlinear | The objective contains $q^2$ |
| Quadratic program (QP) | The objective is quadratic; there are no added constraints |
| Strictly convex | Any stationary point is the unique global minimizer here |

These labels describe different properties. A quadratic objective is not always convex: reversing the sign of this clock score gives a downward-bending quadratic. An LP has affine objective and constraints, and is convex. A nonlinear problem can also be convex, as this example shows.

### 4 · Inspect the curve and its chord

Compare the blue score curve with the dashed line joining the two endpoint scores. Between the endpoints, the curve lies below the line. The teal diamond is the analytically verified minimum. Drawing a finite number of curve points does not replace the convexity proof.


```python
convexity_figure = show_convex_score()
plt.show()
```

### 5 · Practice with worked solutions

**Question 1.** Keep the score, but allow only $q\in\{0,2\}$. Is the feasible set convex?

**Solution.** No. Its two points have midpoint 1, which is not allowed. Convexity of the objective alone does not make the optimization problem convex. Both allowed decisions have score 6 min².

**Question 2.** Does a zero derivative always prove a minimum for a nonlinear objective?

**Solution.** No. The reversed score $-\phi(q)$ also has derivative zero at 1, but it has a maximum there. The convexity assumption matters.
