## 뉴턴 방법 (Newton's Method)

<strong>뉴턴 방법은 일계도함수와 이계도함수를 이용해 갱신량을 결정하는 반복 최적화 방법이다.</strong>

- 일변수 목적함수 $f(x)$를 최소화하는 비제약 문제를 다룬다.

### 1 · 경사하강법 대비 차이점

- 두 방법 모두 현재 위치 $x_k$에서 다음 위치 $x_{k+1}$로 갱신한다. 차이는 <strong>갱신량을 결정하는 정보</strong>에 있다. $k$는 반복 번호이다.
- <strong>경사하강법:</strong> 일계도함수 $f'(x_k)$를 사용하고, 학습률 $\alpha>0$로 갱신량을 조절한다.

$$
x_{k+1}=x_k-\alpha f'(x_k).
$$

- <strong>뉴턴 방법:</strong> 일계도함수에 더해 이계도함수 $f''(x_k)$를 사용한다. 이계도함수는 일계도함수의 변화율이며, 현재 위치의 곡률(curvature)을 반영한다.

$$
x_{k+1}=x_k-\frac{f'(x_k)}{f''(x_k)},
\qquad f''(x_k)\ne0.
$$

- <strong>갱신량의 차이:</strong> 경사하강법은 일계도함수와 학습률로, 뉴턴 방법은 일계도함수와 이계도함수의 비율로 갱신량을 결정한다.
- <strong>곡률의 영향:</strong> 일계도함수 값이 같고 곡률이 양수라면, 곡률이 클수록 뉴턴 갱신량의 크기는 작아진다. 실제 반복에서는 두 도함수가 함께 변하므로, 곡률만으로 갱신량의 변화를 판단할 수 없다.
- <strong>이동 위치의 차이:</strong> 경사하강법은 일계도함수의 반대 방향으로 학습률이 정한 만큼 이동한다. 뉴턴 방법은 곡률이 양수일 때, 현재 위치에서 만든 <strong>국소 이차근사(local quadratic approximation)의 최소점</strong>으로 이동한다.
- 아래 그림은 같은 위치에서 두 방법이 선택한 다음 위치를 비교한다. 파란 실선 $f(x)$는 실제 목적함수, 회색 점선 $q_k(x)$는 현재 위치에서 함수값·일계도함수·이계도함수를 맞춘 이차근사이다.
- 보라색은 경사하강법, 주황색은 뉴턴 방법의 갱신이다. 회색 빈 원은 이차근사의 최소점이며, 주황색 점은 그 위치에서의 실제 함수값이다. 근사의 최소점이 실제 목적함수의 최소점과 같지는 않을 수 있다.

#### 뉴턴 방법의 장점

- <strong>곡률을 반영한 갱신:</strong> 일계도함수의 변화율까지 이용해, 일계도함수가 0이 될 위치를 예측한다. 함수의 국소적인 모양에 맞춰 갱신량을 결정할 수 있다.
- <strong>빠른 국소 수렴:</strong> 함수가 충분히 매끄럽고 최소점의 곡률이 양수인 경우, 최소점 가까이에서 시작하면 적은 반복으로 높은 정확도를 얻을 수 있다.
- <strong>이차함수에서의 정확한 갱신:</strong> 양의 곡률을 갖는 이차함수에서는 근사와 실제 함수가 같으므로, 정확한 도함수를 사용하면 한 번의 갱신으로 최소점에 도달한다.
#### 뉴턴 방법의 단점

- <strong>추가 계산 비용:</strong> 일계도함수뿐 아니라 이계도함수도 계산해야 한다. 반복 횟수가 적어도 전체 계산 비용이 항상 작은 것은 아니다.
- <strong>0에 가까운 곡률:</strong> 곡률이 0이면 뉴턴 식으로 나눌 수 없고, 0에 가까우면 갱신량이 지나치게 커질 수 있다.
- <strong>음의 곡률:</strong> 뉴턴 갱신이 함수값을 증가시키는 방향을 가리킬 수 있다. 이때는 경사하강법으로 전환하고, 함수값이 감소하도록 학습률을 조절하는 것이 좋다.
- <strong>과도한 이동:</strong> 곡률이 양수여도 한 번에 너무 멀리 이동하면 함수값이 증가할 수 있으므로, 실제 함수값을 확인하며 이동량을 조절해야 한다.

### 2 · 곡률이 일정한 경우

- 고정 계수 $a,b,c$와 $a>0$인 이차함수를 생각한다.

$$
f(x)=ax^2+bx+c.
$$

- 기울기는 $f'(x)=2ax+b$, 곡률은 $f''(x)=2a$이다. 곡률이 일정하므로 정확한 도함수로 계산하면 시작 위치와 관계없이 같은 후보를 얻는다.

$$
x_{k+1}=x_k-\frac{2ax_k+b}{2a}=-\frac{b}{2a}.
$$

- 예를 들어 $f(x)=x^2$에서 $x_0=2$이면 기울기는 4, 곡률은 2이다. 따라서 $x_1=2-4/2=0$으로 한 번에 최소점에 도달한다.
- 이차함수는 이차근사와 실제 함수가 같으므로 한 번이면 충분하다. 그림의 $x^*$는 함수값이 가장 작은 <strong>위치</strong>이며, 그때의 <strong>함수값</strong>은 $f(x^*)$이다.

### 3 · 곡률이 변하는 경우

- 일반적인 함수는 위치에 따라 곡률이 달라진다. 현재 위치에서 만든 이차근사는 가까운 곳에서만 실제 함수와 비슷하므로, <strong>한 번 이동해도 실제 최소점에 도달하지 못할 수 있다.</strong>
- 아래 그림에서 근사가 선택한 다음 위치 $x_{k+1}$과 실제 최소점 $x^*$의 차이를 비교한다.

- 예를 들어 $f(x)=\tfrac12x^2+\tfrac14x^4$이면 $f'(x)=x+x^3$, $f''(x)=1+3x^2$이다. $x_0=2$에서 첫 갱신은 다음과 같다.

$$
x_1=2-\frac{2+2^3}{1+3\cdot2^2}
=2-\frac{10}{13}\approx1.231.
$$

- 새 위치에서도 기울기가 0이 아니므로 기울기와 곡률을 다시 계산한다. <strong>목적함수는 그대로 두고, 근사를 만드는 위치를 바꾸며 반복</strong>한다. 아래 두 그림에서 회색 근사와 주황색 갱신 화살표가 달라지는 모습을 비교한다.

### 4 · 뉴턴 방법의 구현과 종료 기준

- $f(x)=\tfrac12x^2+\tfrac14x^4$와 $x_0=2$를 사용한다. 코드에서는 도함수를 근사하는 <strong>수치미분(numerical differentiation)</strong>을 사용한다. 구체적으로는 주변 함수값의 차이를 이용하는 유한차분(finite difference) 방식이다. `derivative`는 기울기, `curvature`는 곡률을 계산한다.
- 수치미분 간격 $h=10^{-4}$는 $x$ 주변의 함수값을 계산하는 작은 간격이다. 뉴턴 방법으로 실제 이동할 양인 `decision_change`와는 다르다. 계산에 근사 오차가 있으므로 기울기가 정확히 0이 되기를 기다리지 않는다.
- 이 예제에서는 허용오차를 $\varepsilon=10^{-6}$으로 정하고, 수치미분으로 계산한 기울기의 크기가 이 값 이하이면 종료한다. $10^{-6}$은 모든 문제에 적용되는 고정 기준이 아니라, 이 예제에서 선택한 설정값이다. 허용오차는 문제의 규모와 단위, 수치미분 오차, 요구 정확도에 따라 조절한다.

$$
|f'(x_k)|\le\varepsilon.
$$

- 기준을 만족하지 않으면 뉴턴 갱신을 반복한다. 무한히 반복하지 않도록 갱신 횟수는 20회로 제한하고, 마지막 갱신 뒤에도 종료 기준을 확인한다.


```python
def objective(x):
    return 0.5 * x**2 + 0.25 * x**4

def derivative(function, x, h=1e-4):
    return (function(x + h) - function(x - h)) / (2 * h)

def curvature(function, x, h=1e-4):
    return (function(x + h) - 2 * function(x) + function(x - h)) / h**2

x = 2.0
tolerance = 1e-6
max_steps = 20
for _ in range(max_steps):
    slope_value = derivative(objective, x)
    if abs(slope_value) <= tolerance:
        break
    curvature_value = curvature(objective, x)
    if curvature_value <= 0:
        raise ValueError("Positive curvature required for Newton's method.")
    x = x -slope_value / curvature_value

converged = abs(derivative(objective, x)) <= tolerance
print("Final variable:", x)
print("Objective value:", objective(x))
print("Converged:", converged)
```
