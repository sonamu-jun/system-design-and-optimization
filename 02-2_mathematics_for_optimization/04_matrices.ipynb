## 02-2 · Part 4: Matrices

**A matrix stores numbers in rows and columns and can transform a vector into another vector.**

The earlier parts introduced vectors and inner products. Matrix multiplication uses one inner product for each output component.

<style>
.katex .textbb { font-style: italic; }
</style>

### 1 · Shape and matrix-vector multiplication

A matrix $A\in\mathbb R^{m\times p}$ has $m$ rows and $p$ columns. The entry $A_{ij}$ lies in row $i$, column $j$, where $i=1,\ldots,m$ and $j=1,\ldots,p$.

Let

$$
A=\begin{bmatrix}2&1\\0&3\end{bmatrix},
\qquad 𝕩=\begin{bmatrix}3\\2\end{bmatrix}.
$$

Take the inner product of each row with the column vector:

$$
A𝕩=\begin{bmatrix}2(3)+1(2)\\0(3)+3(2)\end{bmatrix}
=\begin{bmatrix}8\\6\end{bmatrix}.
$$

In general, an $m\times p$ matrix multiplies a $p$-component column and produces an $m$-component column.

### 2 · Addition, scaling, and transpose

Let $B=\begin{bmatrix}1&0\\0&2\end{bmatrix}$. Add matching entries of matrices with the same shape. Scalar multiplication multiplies every entry:

$$
A+B=\begin{bmatrix}3&1\\0&5\end{bmatrix},
\qquad 2A=\begin{bmatrix}4&2\\0&6\end{bmatrix}.
$$

Transpose exchanges rows and columns:

$$
A^{\mathsf T}=\begin{bmatrix}2&0\\1&3\end{bmatrix}.
$$

For example, $A_{12}=1$ becomes $(A^{\mathsf T})_{21}=1$.

### 3 · Matrix products and identity

For $AB$ to exist, the number of columns of $A$ must equal the number of rows of $B$. Each entry is the inner product of a row of $A$ with a column of $B$:

$$
AB=\begin{bmatrix}2&2\\0&6\end{bmatrix}.
$$

For example, $(AB)_{12}=2(0)+1(2)=2$. The order matters: $BA=\begin{bmatrix}2&1\\0&6\end{bmatrix}$, so $AB\ne BA$ here.

The product combines two transformations: $(AB)𝕩=A(B𝕩)$. The **identity matrix** leaves a vector unchanged:

$$
I_2=\begin{bmatrix}1&0\\0&1\end{bmatrix},
\qquad I_2𝕩=𝕩.
$$
