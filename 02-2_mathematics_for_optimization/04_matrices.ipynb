## 02-2 · Part 4: Matrices

**A matrix calculates several weighted sums of cooling coordinates at once.**

Earlier parts introduced vectors, inner products, and norms. We retain the two cooling coordinates and use inner products to explain matrix operations.

### 1 · Shape and matrix-vector multiplication

A matrix is a rectangular array. Its shape lists rows first, then columns: $A\in\mathbb R^{m\times p}$ has $m$ rows and $p$ columns. The entry $A_{ij}$ is in row $i$ and column $j$, for $i=1,\ldots,m$ and $j=1,\ldots,p$.

For an arithmetic example, let

$$
A=\begin{bmatrix}2&1\\0&3\end{bmatrix},
\qquad 𝕦=\begin{bmatrix}u_{\mathrm{early}}\\u_{\mathrm{late}}\end{bmatrix}
=\begin{bmatrix}3\\2\end{bmatrix}.
$$

The decision sets $u_t=3$ for $t=0,\ldots,5$ and $u_t=2$ for $t=6,\ldots,11$. The matrix $A$ is $2\times2$ and the column $𝕦$ is $2\times1$. Take the inner product of each row with $𝕦$:

$$
A𝕦=\begin{bmatrix}2(3)+1(2)\\0(3)+3(2)\end{bmatrix}
=\begin{bmatrix}8\\6\end{bmatrix}.
$$

The result has one component per row. These are weighted coordinate sums, not temperatures from the classroom system model. In general, an $m\times p$ matrix multiplies a $p$-component column to produce an $m$-component column.

### 2 · Addition, scaling, and transpose

Let $B=\begin{bmatrix}1&0\\0&2\end{bmatrix}$. Add matching entries of matrices with the same shape, and multiply every entry when scaling by a scalar:

$$
A+B=\begin{bmatrix}3&1\\0&5\end{bmatrix},
\qquad 2A=\begin{bmatrix}4&2\\0&6\end{bmatrix}.
$$

Transpose exchanges rows and columns:

$$
A^{\mathsf T}=\begin{bmatrix}2&0\\1&3\end{bmatrix}.
$$

For example, the entry $A_{12}=1$ becomes $(A^{\mathsf T})_{21}=1$. An $m\times p$ matrix becomes a $p\times m$ matrix.

### 3 · Matrix products and order

For $AB$ to exist, the number of columns of $A$ must equal the number of rows of $B$. Each product entry is a row-column inner product. Using the same two matrices,

$$
AB=\begin{bmatrix}2(1)+1(0)&2(0)+1(2)\\0(1)+3(0)&0(0)+3(2)\end{bmatrix}
=\begin{bmatrix}2&2\\0&6\end{bmatrix}.
$$

Matrix multiplication is not entrywise multiplication. Its purpose is to combine successive mappings: $(AB)𝕦=A(B𝕦)$. Here

$$
B𝕦=\begin{bmatrix}3\\4\end{bmatrix},
\qquad A(B𝕦)=\begin{bmatrix}10\\12\end{bmatrix}.
$$

Reversing the order gives $BA=\begin{bmatrix}2&1\\0&6\end{bmatrix}$ and $B(A𝕦)=(8,12)^{\mathsf T}$. The first output differs, so $AB\ne BA$ in this example.

The identity matrix $I_2=\begin{bmatrix}1&0\\0&1\end{bmatrix}$ leaves a two-component vector unchanged: $I_2𝕦=𝕦$.
