## 02-2 · Part 5: Functions with Vector Inputs

**A function assigns exactly one output to each input in its domain.**

Part 4 used matrices to map vectors to vectors. We now study a function that maps a vector to one number.

### 1 · Domain, codomain, and range

Let $𝕩=(x_1,x_2)^{\mathsf T}$. The function

$$
f:\mathbb R^2\longrightarrow\mathbb R,
\qquad f(𝕩)=x_1^2+x_2^2
$$

squares the two components and adds them. For $𝕩=(3,2)^{\mathsf T}$, it gives $f(𝕩)=3^2+2^2=13$.

| Term | Meaning | For this function |
|:---|:---|:---|
| Domain | Allowed inputs | $\mathbb R^2$ |
| Codomain | Stated set containing the outputs | $\mathbb R$ |
| Range | Outputs actually produced | $[0,\infty)$ |

The output cannot be negative. Every nonnegative number $r$ is produced by $(\sqrt r,0)^{\mathsf T}$. Infinity is not an endpoint value, so its interval boundary uses a parenthesis.

Different inputs may share an output: $f(3,2)=f(2,3)=13$.

### 2 · Linear, affine, and quadratic expressions

Let $q$ be a scalar input and $a,b,c$ be fixed coefficients.

| Type | Form | Example |
|:---|:---|:---|
| Linear | $aq$ | $2q$ |
| Affine | $aq+b$ | $2q+1$ |
| Quadratic | $aq^2+bq+c$, $a\ne0$ | $q^2+4$ |

An affine expression with zero offset is linear. A quadratic expression has a nonzero squared term.

In the same function $f$, hold $x_2=2$ and vary $x_1=q$:

$$
f(q,2)=q^2+4.
$$

Inputs $q=1,2,3$ give outputs $5,8,13$. Equal input increases can give different output increases.

### 3 · Composition

**Composition** uses one function's output as another function's input. Let $s:\mathbb R\to\mathbb R$ be $s(z)=2z$. Composing it with $f$ gives

$$
(s\circ f)(𝕩)=s(f(𝕩))=2(x_1^2+x_2^2).
$$

At $𝕩=(3,2)^{\mathsf T}$, first calculate $f(𝕩)=13$, then $s(13)=26$. The symbol $\circ$ means composition. The inner function's output must belong to the outer function's domain.
