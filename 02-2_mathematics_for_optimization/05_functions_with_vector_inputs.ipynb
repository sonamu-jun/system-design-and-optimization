## 02-2 · Part 5: Functions with Vector Inputs

**A function maps the cooling decision vector to one scalar energy output.**

Part 4 mapped a vector to another vector with a matrix. We retain the same cooling decision and now study a function with a vector input and a scalar output. We distinguish its inputs, outputs, coefficients, and composition.

### 1 · Domain, codomain, and range

A function assigns exactly one output to each input in its domain. For $𝕦=(u_{\mathrm{early}},u_{\mathrm{late}})$, the early level supplies $u_0,\ldots,u_5$ and the late level supplies $u_6,\ldots,u_{11}$. The energy function adds the 12 squared cooling actions:

$$
\begin{aligned}
E(𝕦)&=\frac12\sum_{t=0}^{11}u_t^2\\
&=\frac12(6u_{\mathrm{early}}^2+6u_{\mathrm{late}}^2)\\
&=3(u_{\mathrm{early}}^2+u_{\mathrm{late}}^2).
\end{aligned}
$$

We write $E:[0,5]^2\longrightarrow\mathbb R_{\ge0}$, where $\mathbb R_{\ge0}$ is the set of nonnegative real numbers.

| Term | Meaning for $E$ |
|:---|:---|
| Domain | Allowed inputs: $[0,5]^2$ |
| Codomain | Stated set containing outputs: $\mathbb R_{\ge0}$ |
| Range | Outputs actually produced: $[0,150]$ |

The outputs use model energy units. The smallest value is $E(0,0)=0$ and the largest is $E(5,5)=150$. Along $𝕦=(q,q)$ with $0\le q\le5$, the value $6q^2$ covers every value between those endpoints. The requirement $E(𝕦)\le E_{\max}=60$ separately excludes some inputs in the domain.

Different inputs can share an output: $E(3,2)=E(2,3)=3(9+4)=39$. Swapping the cooling periods still changes the schedule; the energy function alone does not distinguish these two schedules.

### 2 · Inputs and fixed coefficients

Let $q$ be a scalar input, and let $k,r,s$ be coefficients held fixed while $q$ changes.

| Type | Expression | Role of the terms |
|:---|:---|:---|
| Linear | $kq$ | A constant multiplier |
| Affine | $kq+r$ | A multiplier and an offset |
| Quadratic | $kq^2+rq+s$, $k\ne0$ | A nonzero squared term |

An affine expression with zero offset is linear. A nonzero offset means doubling the input does not simply double the output.

Fix late cooling at 2 and let only early cooling $q$ vary:

$$
E(q,2)=3q^2+12.
$$

Substituting $q=1,2,3$ gives 15, 24, and 39 model energy units. Equal one-unit input increases produce output increases of 9 and 15. The coefficients stay fixed, but a quadratic function need not have equal output changes for equal input changes.

### 3 · Composition: output becomes input

A composition applies one function to the output of another. The same energy calculation can be separated into a sum of squares and a multiplication:

$$
Q(𝕦)=𝕦^{\mathsf T}𝕦=u_{\mathrm{early}}^2+u_{\mathrm{late}}^2,
\qquad S(z)=3z.
$$

Here $Q$ maps the two-component column $𝕦$ to a nonnegative scalar $z$, and $S$ maps that scalar to energy. The composition symbol $\circ$ gives

$$
E(𝕦)=(S\circ Q)(𝕦)=S(Q(𝕦)).
$$

For $𝕦=(3,2)^{\mathsf T}$, first compute $Q(𝕦)=13$, then $S(13)=39$. The output of the inner function must belong to the domain of the outer function. The order matters: $Q(S(𝕦))$ is not defined with these domains, because $S$ expects a scalar and $𝕦$ is a two-component vector.
