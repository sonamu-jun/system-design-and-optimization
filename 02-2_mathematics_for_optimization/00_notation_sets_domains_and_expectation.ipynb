## 02-2 · Notation: Sets, Domains, and Basic Statistics

**Allowed cooling choices form a set; uncertain weather is described by probabilities and averages.**

Lecture 02-1 used decision domains and weather scenarios. We retain the classroom example and explain the set notation, expectation, and sample statistics behind those expressions.

### 1 · Sets and intervals

A set is a collection of distinct values. $\mathbb R$ contains real numbers; $\mathbb Z$ contains integers. Let $A=\{0,1,\ldots,5\}$ be six allowed cooling levels.

| Symbol | Meaning | Example |
|:---|:---|:---|
| $\in$, $\notin$ | Belongs, does not belong | $2\in A$, $2.5\notin A$ |
| $\subseteq$ | Every member also belongs to another set | $A\subseteq\mathbb R$ |
| $\cap$ | In both sets | $[0,3]\cap[2,5]=[2,3]$ |
| $\cup$ | In either set | $[0,1]\cup[3,5]$ |
| $\varnothing$ | No members | $\{0\}\cap\{1\}=\varnothing$ |

For a real cooling level $q$, square brackets include endpoints; parentheses exclude them:

| Interval | Condition |
|:---|:---|
| $[0,5]$ | $0\le q\le5$ |
| $(0,5)$ | $0<q<5$ |
| $[0,5)$ | $0\le q<5$ |
| $(0,5]$ | $0<q\le5$ |

Set-builder notation $\{q\in\mathbb R:0\le q\le5\}$ means “real values satisfying this condition.” The colon means “such that.”

$[0,1]\cup[3,5]$ has a gap but still contains fractional values such as 3.5. The discrete set $A$ contains only its six listed values.

### 2 · Vectors and domains

The decision $𝕦=(u_{\mathrm{early}},u_{\mathrm{late}})$ chooses two cooling levels over $n=12$ steps:

$$
u_t=\begin{cases}
u_{\mathrm{early}},&t=0,\ldots,5,\\
u_{\mathrm{late}},&t=6,\ldots,11.
\end{cases}
$$

In standard notation, $𝕩$ stores the same ordered coordinates as a column, and $\mathcal X$ is their domain, the set of allowed inputs:

$$
𝕩=\begin{bmatrix}u_{\mathrm{early}}\\u_{\mathrm{late}}\end{bmatrix},
\qquad \mathcal X=[0,5]\times[0,5]\subseteq\mathbb R^2.
$$

Italic double-struck lowercase letters such as $𝕦$ and $𝕩$ denote vectors; scalar coordinates such as $u_{\mathrm{early}}$ and $x_i$ use ordinary type.

The Cartesian product $\times$ forms ordered pairs. $[0,5]^2$ abbreviates this product; it does not mean $[0,25]$.

| Domain | Allowed coordinates |
|:---|:---|
| $[0,5]^2$ | Both continuous |
| $A\times A$ | Both discrete: $6\times6=36$ pairs |
| $A\times[0,5]$ | First discrete, second continuous |

Two coordinates mean membership in $\mathbb R^2$, even when some values must be integers.

$\forall$ means “for every”; $\exists$ means “there exists.” The condition $0\le x_i\le5$ for every $i\in\{1,2\}$ checks both coordinates. For $𝕩=(3,6)$, one coordinate passes, but the condition for every coordinate fails. Additional requirements define a subset $\mathcal F\subseteq\mathcal X$.

### 3 · Probability, population, and sample

Keep the cooling decision fixed. Let the random variable $\xi$ denote a day's outdoor temperature, constant over that day's 12 steps. A random variable assigns a numerical value to an uncertain outcome. Use the three weather scenarios from Lecture 02-1:

| Scenario $s$ | Temperature $\xi_s$ (°C) | Probability $p_s$ |
|:---|:---|:---|
| 1 | 29 | 0.2 |
| 2 | 31 | 0.5 |
| 3 | 33 | 0.3 |

Here $p_s=\mathbb P(\xi=\xi_s)$, with $s=1,2,3$. Probabilities are nonnegative and sum to one. An event is a set of outcomes; for example, $\mathbb P(\xi>30)=0.5+0.3=0.8$.

The **population distribution** specifies the possible values and their probabilities. A **sample** contains observed values drawn from that distribution. For illustration, suppose five observations are 29, 31, 31, 33, and 33 °C.

A population parameter describes the distribution, such as its mean $\mu$. A statistic is calculated from a sample, such as its sample mean. The sample size $m=5$ counts observations; it is not the number of population parameters. These statistical parameters describe the distribution, while $a,b,c$ remain fixed parameters of the classroom system model.

### 4 · Expectation and sample mean

The population mean, or expectation, weights each possible temperature by its probability:

$$
\begin{aligned}
\mu=\mathbb E[\xi]&=\sum_{s=1}^{3}p_s\xi_s\\
&=0.2(29)+0.5(31)+0.3(33)=31.2\;{}^{\circ}\mathrm C.
\end{aligned}
$$

The symbol $\sum$ adds the indexed terms. The mean 31.2 °C need not be one of the possible observations. Averaging the three distinct values equally gives 31 °C, which would assign each scenario probability $1/3$.

Let $\xi^{(k)}$ be observation $k$, for $k=1,\ldots,m$; the parenthesized superscript is an observation index, not a power. The sample mean gives each observation equal weight:

$$
\bar\xi_m=\frac1m\sum_{k=1}^{m}\xi^{(k)},
\qquad \bar\xi_5=\frac{29+31+31+33+33}{5}=31.4\;{}^{\circ}\mathrm C.
$$

Thus the sample mean estimates the population mean, but this sample does not give exactly $\mu=31.2$ °C. Repeated observations of a scenario give that scenario more weight in the sample average.

### 5 · Variance and standard deviation

The mean locates the center; variance measures squared deviations from it. For this distribution, define the population variance $\sigma^2$ and standard deviation $\sigma$ by

$$
\begin{aligned}
\sigma^2=\operatorname{Var}(\xi)&=\sum_{s=1}^{3}p_s(\xi_s-\mu)^2\\
&=0.2(29-31.2)^2+0.5(31-31.2)^2\\
&\quad+0.3(33-31.2)^2=1.96.
\end{aligned}
$$

Variance uses squared-temperature units. Taking the square root gives $\sigma=\sqrt{1.96}=1.4$ °C, in the same unit as temperature. Squaring prevents positive and negative deviations from canceling.

When the population mean is unknown, the usual sample variance uses the sample mean and divides by $m-1$, for $m\ge2$:

$$
s_{\xi}^2=\frac{1}{m-1}\sum_{k=1}^{m}(\xi^{(k)}-\bar\xi_m)^2.
$$

For the five observations above, the squared deviations from 31.4 sum to $5.76+0.16+0.16+2.56+2.56=11.2$. Hence $s_{\xi}^2=11.2/4=2.8$, and the sample standard deviation is about 1.67 °C. The $m-1$ denominator corrects for estimating the mean from the same sample; under independent sampling from one distribution, this estimates the population variance without systematic bias.

### 6 · Why more observations help

Suppose observations are independent draws from this same fixed distribution. Let $m_s$ count observations of scenario $s$, so $m_1+m_2+m_3=m$. Regrouping the sample average gives

$$
\bar\xi_m=\sum_{s=1}^{3}\frac{m_s}{m}\xi_s,
\qquad \mathbb E[\xi]=\sum_{s=1}^{3}p_s\xi_s.
$$

The only difference is the weights: observed frequencies $m_s/m$ versus distribution probabilities $p_s$. As the sample grows, those frequencies approach the probabilities.

For example, a sample of 100 observations with scenario counts 21, 49, and 30 has mean

$$
\bar\xi_{100}=\frac{21(29)+49(31)+30(33)}{100}=31.18\;{}^{\circ}\mathrm C.
$$

This particular sample is closer to the population mean of 31.2 °C than the five-observation sample.

The **law of large numbers** states that, for independent observations from the same distribution with a finite expectation, the sample mean converges to that expectation as $m$ grows. In symbols,

$$
\bar\xi_m\longrightarrow\mu=\mathbb E[\xi]\qquad\text{as }m\to\infty.
$$

This does not promise exact equality for a finite sample.

Nor must every added observation improve the estimate. If the sixth temperature is 33 °C, then $\bar\xi_6=(157+33)/6\approx31.67$ °C, farther from 31.2 than the five-observation mean. A larger sample also does not repair a sample drawn from the wrong distribution, such as observing only unusually hot days.

### 7 · Expectation of a performance score

The same weighting applies to the classroom score. Retain $𝕦=(3,2)$ and the energy-weight hyperparameter $\lambda_E=1$. Discomfort $D(𝕦)$ and energy $E(𝕦)$ are performance outputs; the score mapping $H$ combines them:

$$
\begin{aligned}
J(𝕦;\lambda_E)&=H(D(𝕦),E(𝕦);\lambda_E)\\
&=D(𝕦)+\lambda_E E(𝕦).
\end{aligned}
$$

Keep the initial indoor temperature $T_0=27$ °C, occupancy $N_t=20$, and fixed parameters $(a,b,c)=(0.12,0.012,0.45)$. Change only the constant outdoor temperature $T_t^{\mathrm{out}}$ between scenarios. The resulting rounded scores $J_s$ from the classroom system are

| Outdoor temperature (°C) | Probability $p_s$ | Score $J_s$ |
|:---|:---|:---|
| 29 | 0.2 | 45.885 |
| 31 | 0.5 | 53.465 |
| 33 | 0.3 | 86.842 |

$$
\begin{aligned}
\mathbb E_{\xi}[J]&=\sum_{s=1}^{3}p_sJ_s\\
&=0.2(45.885)+0.5(53.465)+0.3(86.842)\\
&\approx61.962.
\end{aligned}
$$

An average of independently sampled scores approaches this expected score under the same sampling assumptions. The expectation operator $\mathbb E[\cdot]$ differs from the energy output $E(𝕦)$.

Evaluate the score for each weather outcome before averaging: the score at the mean temperature need not equal the mean score. Expectation alone also does not show that every scenario satisfies the requirements.
