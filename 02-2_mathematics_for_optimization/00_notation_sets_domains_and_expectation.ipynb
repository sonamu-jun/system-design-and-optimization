## 02-2 · Notation: Sets, Domains, and Basic Statistics

**Sets describe collections of numbers; probability and statistics describe uncertain values.**

Lecture 02-1 used these symbols in problem formulations. Here we explain them with small numerical examples.

### 1 · Sets and intervals

A **set** is a collection of distinct elements. $\mathbb R$ is the set of real numbers, and $\mathbb Z$ is the set of integers. Order does not matter: $\{1,2\}=\{2,1\}$.

| Notation | Meaning |
|:---|:---|
| $A=\{0,1,2\}$ | A set with three elements |
| $1\in A$, $3\notin A$ | 1 belongs to $A$; 3 does not |
| $A\subseteq\mathbb R$ | Every element of $A$ is real |
| $A\cap\{2,3\}=\{2\}$ | Intersection: elements in both sets |
| $A\cup\{2,3\}=\{0,1,2,3\}$ | Union: elements in either set |
| $A\cap\{3\}=\varnothing$ | The empty set has no elements |

An **interval** contains every real number between its endpoints. A square bracket includes an endpoint; a parenthesis excludes it.

| Interval | Condition |
|:---|:---|
| $[0,2]$ | $0\le x\le2$ |
| $(0,2)$ | $0<x<2$ |
| $[0,2)$ | $0\le x<2$ |
| $(0,2]$ | $0<x\le2$ |

For example, $1.5\in[0,2]$ but $1.5\notin A$. A set can also be defined by a condition:

$$
[0,2]=\{x\in\mathbb R:0\le x\le2\}.
$$

The colon means “such that.”

### 2 · Ordered pairs and domains

An ordered pair keeps two coordinates in a fixed order. The Cartesian product $\times$ forms all pairs from two sets:

$$
\{0,1\}\times\{2,3\}=\{(0,2),(0,3),(1,2),(1,3)\}.
$$

$\mathbb R^2$ is the set of all ordered pairs of real numbers. A **domain** is a set of allowed inputs. For example,

$$
\mathcal X=[0,2]^2=[0,2]\times[0,2].
$$

Both coordinates must lie in $[0,2]$: $(1,2)\in\mathcal X$, but $(1,3)\notin\mathcal X$. The superscript 2 counts coordinates.

For coordinates $x_1=1$ and $x_2=3$, compare the quantifiers:

| Symbol | Statement | Result |
|:---|:---|:---|
| $\forall$: for every | $\forall i\in\{1,2\},\;x_i\le2$ | False: $x_2=3$ |
| $\exists$: there exists | $\exists i\in\{1,2\}:\;x_i\le2$ | True: $x_1=1$ |

### 3 · Probability and expectation

A **random variable** assigns a number to an uncertain outcome. Let $X$ have the following distribution, which lists its possible values and their probabilities:

| Value $x_i$ | Probability $p_i=\mathbb P(X=x_i)$ |
|:---|:---|
| 0 | $1/4$ |
| 1 | $1/2$ |
| 2 | $1/4$ |

Here $i=1,2,3$ indexes the rows. Probabilities are nonnegative and sum to 1. For example, $\mathbb P(X\ge1)=1/2+1/4=3/4$.

The **expectation**, or population mean $\mu$, is a probability-weighted average:

$$
\mu=\mathbb E[X]=\sum_{i=1}^{3}p_ix_i
=\frac14(0)+\frac12(1)+\frac14(2)=1.
$$

The symbol $\sum$ adds the terms over the stated index range.

### 4 · Variance and standard deviation

**Variance** measures squared distance from the mean. For the same distribution with $\mu=1$,

$$
\begin{aligned}
\sigma^2=\operatorname{Var}(X)&=\sum_{i=1}^{3}p_i(x_i-\mu)^2\\
&=\frac14(0-1)^2+\frac12(1-1)^2+\frac14(2-1)^2\\
&=\frac12.
\end{aligned}
$$

Squaring prevents positive and negative deviations from canceling. The **standard deviation** is $\sigma=\sqrt{\sigma^2}=\sqrt{1/2}\approx0.707$.

### 5 · Sample mean and sample variance

A **sample** contains observed values. The distribution above has mean $\mu=1$, but a sample need not have that mean.

Let $z_k$ denote observation $k$, for $k=1,\ldots,m$, where $m$ is the sample size. For the sample $0,1,1$, we have $m=3$ and sample mean

$$
\bar z=\frac1m\sum_{k=1}^{m}z_k=\frac{0+1+1}{3}=\frac23.
$$

The usual sample variance uses $m-1$ in the denominator, for $m\ge2$:

$$
s^2=\frac{1}{m-1}\sum_{k=1}^{m}(z_k-\bar z)^2
=\frac{4/9+1/9+1/9}{2}=\frac13.
$$

The $m-1$ denominator accounts for estimating the mean from the same sample. The quantities $\mu$ and $\sigma^2$ describe the distribution; $\bar z$ and $s^2$ are statistics calculated from observations.
