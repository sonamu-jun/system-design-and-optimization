## 02-2 · Part 7: Vector Differentiation and the Chain Rule

**Vector differentiation tracks how a cooling change passes through several outputs to the final score.**

Part 6 calculated the gradient of the scalar energy output. We retain the cooling vector and distinguish derivatives of vector outputs from gradients of scalar outputs, then calculate the classroom score gradient.

### 1 · A vector can depend on a scalar

Keep the starting decision $\boldsymbol{u}=(3,2)^{\mathsf T}$ and displacement $\boldsymbol{d}=(0.5,-0.5)^{\mathsf T}$. Define a family of decisions with scalar multiplier $\alpha$:

$$
\boldsymbol{v}(\alpha)=\boldsymbol{u}+\alpha\boldsymbol{d}
=\begin{bmatrix}3+0.5\alpha\\2-0.5\alpha\end{bmatrix}.
$$

The derivative of a vector with respect to a scalar differentiates each component:

$$
\frac{\mathrm d\boldsymbol{v}}{\mathrm d\alpha}
=\begin{bmatrix}0.5\\-0.5\end{bmatrix}=\boldsymbol{d}.
$$

This derivative is a vector. At $\alpha=0$, the energy change per unit change in $\alpha$ is instead a scalar:

$$
\left.\frac{\mathrm dE(\boldsymbol{v}(\alpha))}{\mathrm d\alpha}\right|_{\alpha=0}
=\begin{bmatrix}18&12\end{bmatrix}\boldsymbol{d}=3.
$$

Direct substitution confirms it: $E(\boldsymbol{v}(\alpha))=39+3\alpha+1.5\alpha^2$, whose derivative at zero is 3. The input and output types determine the derivative's shape.

### 2 · Useful gradients of vector expressions

Let $\boldsymbol{x}\in\mathbb R^p$ be a column vector. Hold the vectors $\boldsymbol{v}\in\mathbb R^p$, $\boldsymbol{z}\in\mathbb R^q$ and matrices $M\in\mathbb R^{p\times p}$, $A\in\mathbb R^{q\times p}$ fixed. Each expression below is scalar, so its gradient has $p$ components.

| Scalar expression | Gradient with respect to $\boldsymbol{x}$ |
|:---|:---|
| $\boldsymbol{v}^{\mathsf T}\boldsymbol{x}$ | $\boldsymbol{v}$ |
| $\boldsymbol{x}^{\mathsf T}\boldsymbol{x}=\lVert\boldsymbol{x}\rVert_2^2$ | $2\boldsymbol{x}$ |
| $\boldsymbol{x}^{\mathsf T}M\boldsymbol{x}$ | $(M+M^{\mathsf T})\boldsymbol{x}$ |
| $\tfrac12\lVert A\boldsymbol{x}-\boldsymbol{z}\rVert_2^2$ | $A^{\mathsf T}(A\boldsymbol{x}-\boldsymbol{z})$ |

For the energy function, take $\boldsymbol{x}=\boldsymbol{u}$ and $M=3I_2$, where $I_2$ is the identity matrix. The quadratic rule gives $(3I_2+3I_2)\boldsymbol{u}=6\boldsymbol{u}$, matching the component calculation.

The shortcut $2M\boldsymbol{x}$ requires $M=M^{\mathsf T}$, meaning that $M$ is symmetric. To check why, use the arithmetic example

$$
M=\begin{bmatrix}2&1\\0&3\end{bmatrix},
\qquad \boldsymbol{x}^{\mathsf T}M\boldsymbol{x}=2x_1^2+x_1x_2+3x_2^2.
$$

Differentiating the expanded expression gives $(4x_1+x_2,x_1+6x_2)^{\mathsf T}$. At the same coordinate values $\boldsymbol{x}=(3,2)^{\mathsf T}$, this is $(14,15)^{\mathsf T}$, equal to $(M+M^{\mathsf T})\boldsymbol{x}$. The expression $2M\boldsymbol{x}$ would give $(16,12)^{\mathsf T}$. This arithmetic expression is separate from the physical energy function.

### 3 · A Jacobian collects derivatives of vector outputs

For a mapping from $\boldsymbol{x}\in\mathbb R^p$ to $\boldsymbol{r}(\boldsymbol{x})\in\mathbb R^q$, define the Jacobian matrix $\mathcal A$ by

$$
\mathcal A_{ij}(\boldsymbol{x})=\frac{\partial r_i}{\partial x_j},
\qquad i=1,\ldots,q,\quad j=1,\ldots,p.
$$

Rows identify outputs and columns identify inputs. Thus $\mathcal A$ has shape $q\times p$. Each row is the transpose of one scalar output's gradient.

For the matrix mapping from Part 4,

$$
\boldsymbol{r}(\boldsymbol{x})
=\begin{bmatrix}2x_1+x_2\\3x_2\end{bmatrix},
\qquad \mathcal A=\begin{bmatrix}2&1\\0&3\end{bmatrix}=A.
$$

For $\Delta\boldsymbol{x}=(0.01,-0.01)^{\mathsf T}$, the output change is $A\Delta\boldsymbol{x}=(0.01,-0.03)^{\mathsf T}$. This is exact because the mapping is linear. For a differentiable nonlinear mapping, $\mathcal A(\boldsymbol{x})\Delta\boldsymbol{x}$ is generally a local approximation.

| Mapping | Derivative representation | Shape |
|:---|:---|:---|
| Scalar $\alpha$ to vector $\boldsymbol{x}(\alpha)$ | Component derivatives | $p\times1$ |
| Vector $\boldsymbol{x}$ to a scalar | Column gradient | $p\times1$ |
| Vector $\boldsymbol{x}$ to vector $\boldsymbol{r}$ | Jacobian | $q\times p$ |

With this convention, a scalar output's Jacobian is its gradient transpose. We reserve $J$ for the classroom score and use $\mathcal A$ for the Jacobian.

### 4 · The chain rule combines successive derivatives

Part 5 wrote energy as $E(\boldsymbol{u})=S(Q(\boldsymbol{u}))$, where $Q(\boldsymbol{u})=\boldsymbol{u}^{\mathsf T}\boldsymbol{u}$ and $S(z)=3z$. The chain rule multiplies the outer derivative by the inner gradient:

$$
\nabla_{\boldsymbol{u}}E
=S'(Q(\boldsymbol{u}))\nabla_{\boldsymbol{u}}Q
=3(2\boldsymbol{u})=6\boldsymbol{u}.
$$

For the classroom, write the two performance outputs as $\boldsymbol{y}=\operatorname{Sim}(\boldsymbol{x})=(D(\boldsymbol{x}),E(\boldsymbol{x}))^{\mathsf T}$, with $\boldsymbol{x}=\boldsymbol{u}$. The simulation composes repeated physical state transitions $F$ with the performance mapping $G$.

The score mapping $H$ supplies the scalar objective $f(\boldsymbol{y};\lambda_E)=y_1+\lambda_Ey_2$ in standard notation:

$$
\begin{aligned}
J(\boldsymbol{u};\lambda_E)&=H(D(\boldsymbol{u}),E(\boldsymbol{u});\lambda_E)\\
&=D(\boldsymbol{u})+\lambda_E E(\boldsymbol{u}).
\end{aligned}
$$

Let $\mathcal A$ now be the Jacobian of this performance mapping. Holding the energy-weight hyperparameter $\lambda_E$ fixed gives

$$
\begin{aligned}
\nabla_{\boldsymbol{u}}J
&=\mathcal A^{\mathsf T}\nabla_{\boldsymbol{y}}f
=\mathcal A^{\mathsf T}\begin{bmatrix}1\\\lambda_E\end{bmatrix}\\
&=\nabla_{\boldsymbol{u}}D+\lambda_E\nabla_{\boldsymbol{u}}E.
\end{aligned}
$$

The transpose combines the two output gradients into one decision gradient. Calculating it requires the effect of cooling on discomfort through the temperature path.

### 5 · Differentiate the physical temperature transition

The decision $\boldsymbol{u}=(u_{\mathrm{early}},u_{\mathrm{late}})^{\mathsf T}$ expands to $u_t=u_{\mathrm{early}}$ for $t=0,\ldots,5$ and $u_t=u_{\mathrm{late}}$ for $t=6,\ldots,11$. The horizon has $n=12$ actions and states $T_0,\ldots,T_{12}$.

Retain the classroom transition and initial condition:

$$
\begin{aligned}
T_{t+1}&=F(T_t,T_t^{\mathrm{out}},N_t,u_t;a,b,c)\\
&=T_t+a(T_t^{\mathrm{out}}-T_t)+bN_t-cu_t,\\
T_0&=27\;{}^{\circ}\mathrm C,\qquad t=0,\ldots,11.
\end{aligned}
$$

Hold the external inputs $T_t^{\mathrm{out}}=31$ °C and $N_t=20$ fixed, with fixed parameters $(a,b,c)=(0.12,0.012,0.45)$.

Define the sensitivity vector $\boldsymbol{s}_t=\nabla_{\boldsymbol{u}}T_t$. Its components measure how the same state responds to the two cooling settings. Define the coordinate vectors $\boldsymbol{e}_{\mathrm{early}}=(1,0)^{\mathsf T}$ and $\boldsymbol{e}_{\mathrm{late}}=(0,1)^{\mathsf T}$. Let $\boldsymbol{\eta}_t$ select the early vector for $t=0,\ldots,5$ and the late vector for $t=6,\ldots,11$.

The initial state does not depend on the decision. Differentiating the transition gives

$$
\boldsymbol{s}_0=\boldsymbol{0},
\qquad \boldsymbol{s}_{t+1}=0.88\boldsymbol{s}_t-0.45\boldsymbol{\eta}_t,
\qquad t=0,\ldots,11.
$$

For example, $\boldsymbol{s}_1=(-0.45,0)^{\mathsf T}$ and $\boldsymbol{s}_2=0.88\boldsymbol{s}_1-0.45\boldsymbol{e}_{\mathrm{early}}=(-0.846,0)^{\mathsf T}$. Late cooling cannot affect these early states. Repeating the same calculation gives

| State | Early sensitivity (°C per cooling unit) | Late sensitivity (°C per cooling unit) |
|:---|:---|:---|
| $T_6$ | $-2.008485$ | 0 |
| $T_7$ | $-1.767467$ | $-0.45$ |
| $T_{12}$ | $-0.932748$ | $-2.008485$ |

Early cooling still affects later temperatures because the state is carried forward.

### 6 · Calculate the score gradient

Discomfort adds a penalty $\phi$ for each state outside the 22–24 °C comfort range:

$$
\phi(T)=\max(T-24,0)^2+\max(22-T,0)^2,
\qquad D(\boldsymbol{u})=\sum_{t=1}^{12}\phi(T_t).
$$

Inside the comfort range the penalty is zero. Outside it, differentiate the corresponding square. The slopes agree at both endpoints:

$$
\phi'(T)=\begin{cases}
2(T-22),&T<22,\\
0,&22\le T\le24,\\
2(T-24),&T>24.
\end{cases}
$$

Apply the chain rule to each state and add:

$$
\nabla_{\boldsymbol{u}}D
=\sum_{t=1}^{12}\phi'(T_t)\boldsymbol{s}_t.
$$

At $\boldsymbol{u}=(3,2)^{\mathsf T}$, the first state is $T_1=26.37$ °C, so its contribution is $4.74(-0.45,0)^{\mathsf T}=(-2.133,0)^{\mathsf T}$. Adding all twelve contributions gives

$$
\nabla_{\boldsymbol{u}}D\approx\begin{bmatrix}-23.152108\\-11.158153\end{bmatrix},
\qquad \nabla_{\boldsymbol{u}}E=\begin{bmatrix}18\\12\end{bmatrix}.
$$

Therefore, with $\lambda_E=1$,

$$
\nabla_{\boldsymbol{u}}J
=\nabla_{\boldsymbol{u}}D+\nabla_{\boldsymbol{u}}E
\approx\begin{bmatrix}-5.152108\\0.841847\end{bmatrix}.
$$

A small increase in early cooling reduces the score locally: its discomfort reduction outweighs its energy cost. A small increase in late cooling raises the score locally at this decision and weight. The components use score units per cooling unit.

Changing only $\lambda_E$ changes the combination of the two performance gradients, while the temperature path and performance gradients stay fixed. A proposed decision still requires separate checks against $u_{\min}=0$, $u_{\max}=5$, the state limits $T_{\min}=20$ °C and $T_{\max}=30$ °C, and $E_{\max}=60$. A local slope is not an acceptance test.
