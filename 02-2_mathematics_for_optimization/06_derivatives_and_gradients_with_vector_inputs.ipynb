## 02-2 · Part 6: Derivatives and Gradients with Vector Inputs

**Differentiating energy with respect to the cooling vector measures the effect of each cooling setting.**

Part 5 defined a scalar energy function with the vector input $𝕦$. We retain that function, begin with a one-variable derivative, and then collect partial derivatives with respect to the vector’s components into a gradient. We then prove the derivative rules for linear expressions, squared norms, quadratic forms, and squared errors.

.katex .textbb { font-style: italic; }
</style>

### 1 · Change one input and measure the output change

The decision $𝕦=(u_{\mathrm{early}},u_{\mathrm{late}})^{\mathsf T}$ applies $u_{\mathrm{early}}$ at $t=0,\ldots,5$ and $u_{\mathrm{late}}$ at $t=6,\ldots,11$. Its energy output is

$$
E(𝕦)=3(u_{\mathrm{early}}^2+u_{\mathrm{late}}^2).
$$

Hold late cooling at 2. Let $q$ denote early cooling and define the scalar function $e(q)=E(q,2)=3q^2+12$. At $q=3$, energy is $e(3)=39$ model energy units.

Increase $q$ by a nonzero amount $\varepsilon$. The quotient

$$
\frac{e(q+\varepsilon)-e(q)}{\varepsilon}
$$

is the average energy change per cooling unit over that interval. For $\varepsilon=0.1$, the energy change is $e(3.1)-e(3)=1.83$, giving an average rate of 18.3. For $\varepsilon=0.01$, the change is 0.1803 and the rate is 18.03.

### 2 · A derivative is the limiting rate

The derivative $e'(q)$, also written $\mathrm de/\mathrm dq$, is the limit of those rates as the input change tends to zero:

$$
e'(q)=\lim_{\varepsilon\to0}\frac{e(q+\varepsilon)-e(q)}{\varepsilon}.
$$

Calculate the quotient before taking the limit:

$$
\begin{aligned}
\frac{e(q+\varepsilon)-e(q)}{\varepsilon}
&=\frac{3(q+\varepsilon)^2-3q^2}{\varepsilon}\\
&=\frac{6q\varepsilon+3\varepsilon^2}{\varepsilon}
=6q+3\varepsilon.
\end{aligned}
$$

Therefore $e'(q)=6q$ and $e'(3)=18$ energy units per cooling unit. The prime means derivative, not a power. A small positive change in early cooling increases energy at this local rate.

For fixed scalar constants $k$ and $r$, the basic rules used here are

| Function of $q$ | Derivative |
|:---|:---|
| $r$ | $0$ |
| $kq$ | $k$ |
| $kq^2$ | $2kq$ |

Differentiate a sum term by term. A constant term contributes zero because it does not change with the input.

### 3 · Partial derivatives hold the other input fixed

The full energy function has two inputs. A partial derivative changes one coordinate while holding the other fixed. The symbol $\partial$ indicates this coordinate-wise derivative.

For early cooling, the late-cooling term is constant:

$$
\frac{\partial E}{\partial u_{\mathrm{early}}}
=6u_{\mathrm{early}}+0=6u_{\mathrm{early}}.
$$

For late cooling, the early-cooling term is constant:

$$
\frac{\partial E}{\partial u_{\mathrm{late}}}
=0+6u_{\mathrm{late}}=6u_{\mathrm{late}}.
$$

At $𝕦=(3,2)^{\mathsf T}$, these rates are 18 and 12. The same small increase in early cooling has a larger energy effect at this decision. Both derivatives describe the same function at the same decision; only the coordinate being changed differs.

### 4 · Collect the partial derivatives into a gradient

For a differentiable scalar function, the **gradient** is the column vector of partial derivatives. Keep the same coordinate order as the input vector:

$$
\nabla_{𝕦}E(𝕦)
=\begin{bmatrix}
\partial E/\partial u_{\mathrm{early}}\\
\partial E/\partial u_{\mathrm{late}}
\end{bmatrix}
=\begin{bmatrix}6u_{\mathrm{early}}\\6u_{\mathrm{late}}\end{bmatrix}
=6𝕦.
$$

The symbol $\nabla$ is read “nabla”; its subscript states which vector is differentiated. Substituting the current decision gives

$$
\left.\nabla_{𝕦}E\right|_{𝕦=(3,2)^{\mathsf T}}
=\begin{bmatrix}18\\12\end{bmatrix}.
$$

The vertical bar means “evaluated at.” The energy value is the scalar 39; its gradient has two components describing sensitivity to the two cooling settings. The gradient has the same number of components as the decision vector.

### 5 · Differentiate a linear expression

The energy calculation used squared cooling coordinates. To compare linear and quadratic dependence, write the same coordinates as a general column $𝕩=(x_1,\ldots,x_p)^{\mathsf T}\in\mathbb R^p$, where $p$ is the number of components. In the classroom example, $p=2$ and $𝕩=𝕦$. The following arithmetic expressions use these coordinates without changing the physical energy function.

Let $𝕨=(w_1,\ldots,w_p)^{\mathsf T}$ and the scalar $r$ be fixed. A weighted sum with an offset is affine; with $r=0$ it is linear:

$$
f_L(𝕩)=𝕨^{\mathsf T}𝕩+r=\sum_{i=1}^{p}w_ix_i+r.
$$

**Proof.** For each $k=1,\ldots,p$, only $w_kx_k$ changes when $x_k$ changes. Every other term is constant, so

$$
\frac{\partial f_L}{\partial x_k}=w_k,
\qquad \nabla_{𝕩}f_L=𝕨.
$$

The scalar expressions $𝕩^{\mathsf T}𝕨$ and $𝕨^{\mathsf T}𝕩$ are equal, so both have gradient $𝕨$. We use a column gradient throughout.

For $𝕨=(2,1)^{\mathsf T}$ and $r=0$, the expression is $f_L(𝕦)=2u_{\mathrm{early}}+u_{\mathrm{late}}$. At $𝕦=(3,2)^{\mathsf T}$, its value is 8 and its gradient is $(2,1)^{\mathsf T}$. The gradient is constant because the coefficients are fixed.

### 6 · Differentiate a squared norm

Now each coordinate appears squared. The squared Euclidean norm is the scalar function

$$
f_N(𝕩)=𝕩^{\mathsf T}𝕩=\lVert𝕩\rVert_2^2=\sum_{i=1}^{p}x_i^2.
$$

**Proof.** Holding all other coordinates fixed leaves just $x_k^2$ to differentiate:

$$
\frac{\partial f_N}{\partial x_k}=2x_k,
\qquad k=1,\ldots,p.
$$

Collecting the components gives

$$
\nabla_{𝕩}(𝕩^{\mathsf T}𝕩)=2𝕩,
\qquad
\nabla_{𝕩}\left(\frac12𝕩^{\mathsf T}𝕩\right)=𝕩.
$$

Unlike a linear expression, this gradient depends on the input. The rule is for the **squared norm**; it is not the derivative of $\lVert𝕩\rVert_2$ itself.

The classroom energy is $E(𝕦)=3𝕦^{\mathsf T}𝕦$. Multiplying the rule by 3 gives $\nabla_{𝕦}E=6𝕦=(18,12)^{\mathsf T}$ at $𝕦=(3,2)^{\mathsf T}$, matching the earlier partial derivatives.

### 7 · Differentiate a general quadratic form

A fixed matrix $M\in\mathbb R^{p\times p}$ allows products of different coordinates as well as squares. The resulting scalar expression is called a quadratic form:

$$
f_Q(𝕩)=𝕩^{\mathsf T}M𝕩
=\sum_{i=1}^{p}\sum_{j=1}^{p}M_{ij}x_ix_j.
$$

Here $M_{ij}$ is the entry in row $i$, column $j$. Keep every entry fixed while differentiating.

**Proof.** Fix $k\in\{1,\ldots,p\}$. The terms containing $x_k$ are $M_{kk}x_k^2$, $M_{kj}x_kx_j$ for $j\ne k$, and $M_{ik}x_ix_k$ for $i\ne k$. Therefore

$$
\begin{aligned}
\frac{\partial f_Q}{\partial x_k}
&=2M_{kk}x_k+\sum_{j\ne k}M_{kj}x_j
+\sum_{i\ne k}M_{ik}x_i\\
&=\sum_{j=1}^{p}M_{kj}x_j+\sum_{i=1}^{p}M_{ik}x_i\\
&=(M𝕩)_k+(M^{\mathsf T}𝕩)_k.
\end{aligned}
$$

The restricted sums run over indices from 1 to $p$, excluding $k$. A subscript $k$ on a vector expression selects component $k$. Collecting all components proves

$$
\nabla_{𝕩}(𝕩^{\mathsf T}M𝕩)=(M+M^{\mathsf T})𝕩.
$$

The shortcut $2M𝕩$ applies when $M$ is **symmetric**, meaning $M=M^{\mathsf T}$. Under that condition,

$$
\nabla_{𝕩}(𝕩^{\mathsf T}M𝕩)=2M𝕩,
\qquad
\nabla_{𝕩}\left(\frac12𝕩^{\mathsf T}M𝕩\right)=M𝕩.
$$

For energy, $M=3I_2$ is symmetric, where $I_2$ is the two-dimensional identity matrix. The result is again $6𝕦$.

To check why symmetry matters, use the same matrix entries as in Part 4 and the same input $𝕩=𝕦=(3,2)^{\mathsf T}$:

$$
M=\begin{bmatrix}2&1\\0&3\end{bmatrix},
\qquad f_Q(𝕩)=2x_1^2+x_1x_2+3x_2^2.
$$

Direct differentiation gives

$$
\nabla_{𝕩}f_Q
=\begin{bmatrix}4x_1+x_2\\x_1+6x_2\end{bmatrix}
=\begin{bmatrix}14\\15\end{bmatrix}
=(M+M^{\mathsf T})𝕩.
$$

At this same matrix and input, $2M𝕩=(16,12)^{\mathsf T}$ would be incorrect.

Adding the linear and constant terms uses the sum rule. For fixed symmetric $M$, fixed $𝕨$, and fixed $r$,

$$
\nabla_{𝕩}\left(\frac12𝕩^{\mathsf T}M𝕩+𝕨^{\mathsf T}𝕩+r\right)
=M𝕩+𝕨.
$$

Here “quadratic” describes the degree of the input expression. All these calculations take one derivative.

### 8 · Differentiate a squared error

Suppose the weighted coordinate sums $A𝕩$ are compared with a fixed target vector $𝕫$. Let $A\in\mathbb R^{m\times p}$ and $𝕫\in\mathbb R^m$ be fixed, where $m$ is the number of sums. Half the sum of squared differences is

$$
f_R(𝕩)=\frac12\lVert A𝕩-𝕫\rVert_2^2.
$$

**Proof.** Expand the inner product. The two cross terms are equal scalars:

$$
\begin{aligned}
f_R(𝕩)
&=\frac12(A𝕩-𝕫)^{\mathsf T}(A𝕩-𝕫)\\
&=\frac12𝕩^{\mathsf T}A^{\mathsf T}A𝕩
-(A^{\mathsf T}𝕫)^{\mathsf T}𝕩+\frac12𝕫^{\mathsf T}𝕫.
\end{aligned}
$$

The matrix $A^{\mathsf T}A$ is symmetric because $(A^{\mathsf T}A)^{\mathsf T}=A^{\mathsf T}A$. Apply the quadratic and linear rules; the last term is constant:

$$
\nabla_{𝕩}f_R=A^{\mathsf T}A𝕩-A^{\mathsf T}𝕫
=A^{\mathsf T}(A𝕩-𝕫).
$$

Using the Part 4 matrix $A=\begin{bmatrix}2&1\\0&3\end{bmatrix}$, $𝕩=𝕦=(3,2)^{\mathsf T}$, and $𝕫=(6,6)^{\mathsf T}$ gives

$$
A𝕩-𝕫=\begin{bmatrix}2\\0\end{bmatrix},
\qquad f_R(𝕩)=2,
\qquad \nabla_{𝕩}f_R
=\begin{bmatrix}2&0\\1&3\end{bmatrix}\begin{bmatrix}2\\0\end{bmatrix}
=\begin{bmatrix}4\\2\end{bmatrix}.
$$

This expression measures error in the weighted sums. Its gradient has $p$ components, one for each input coordinate. Omitting the factor $1/2$ doubles both the function and its gradient.

### 9 · Use the gradient to estimate a combined change

When both coordinates change, the inner product combines their local effects. For a small displacement $\Delta𝕦$,

$$
E(𝕦+\Delta𝕦)-E(𝕦)
\approx[\nabla_{𝕦}E(𝕦)]^{\mathsf T}\Delta𝕦.
$$

Take a small multiple of the displacement from Part 1: $\Delta𝕦=0.02𝕕=(0.01,-0.01)^{\mathsf T}$. At $𝕦=(3,2)^{\mathsf T}$, the predicted energy change is

$$
18(0.01)+12(-0.01)=0.06.
$$

The exact change is $E(3.01,1.99)-E(3,2)=39.0606-39=0.0606$. The prediction omits the quadratic correction:

$$
\begin{aligned}
E(𝕦+\Delta𝕦)-E(𝕦)
&=(6𝕦)^{\mathsf T}\Delta𝕦\\
&\quad+3\lVert\Delta𝕦\rVert_2^2.
\end{aligned}
$$

Here the correction is $3(0.01^2+(-0.01)^2)=0.0006$. The gradient supplies the linear part of the change, so it need not give the exact change over a finite displacement. Energy sensitivity alone does not describe discomfort or show that a new schedule meets the requirements.
