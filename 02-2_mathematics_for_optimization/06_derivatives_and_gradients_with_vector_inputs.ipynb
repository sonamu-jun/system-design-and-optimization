## 02-2 · Part 6: Derivatives and Gradients with Vector Inputs

**A derivative measures a local rate of change; a gradient collects the rates for all input coordinates.**

Part 5 introduced $f(𝕩)=x_1^2+x_2^2$. We keep this function and progress from one-variable derivatives to vector differentiation.

### 1 · A derivative is a limiting rate

Hold $x_2=2$ and write $q=x_1$. Then $e(q)=f(q,2)=q^2+4$. For a nonzero input change $h$, the average rate of change is

$$
\frac{e(q+h)-e(q)}{h}
=\frac{(q+h)^2-q^2}{h}=2q+h.
$$

The **derivative** is the limit as $h$ tends to zero:

$$
e'(q)=\lim_{h\to0}\frac{e(q+h)-e(q)}{h}=2q.
$$

At $q=3$, the average rate is 6.1 for $h=0.1$ and 6.01 for $h=0.01$. These approach $e'(3)=6$. The prime indicates differentiation.

For fixed constants $a,b$, the basic rules are:

| Function | Derivative |
|:---|:---|
| $b$ | $0$ |
| $aq$ | $a$ |
| $aq^2$ | $2aq$ |

Differentiate sums term by term. A constant contributes zero.

### 2 · Partial derivatives

A **partial derivative** changes one coordinate while holding the others fixed. For $f(𝕩)=x_1^2+x_2^2$, the symbol $\partial$ indicates which coordinate changes:

$$
\frac{\partial f}{\partial x_1}=2x_1,
\qquad \frac{\partial f}{\partial x_2}=2x_2.
$$

When differentiating with respect to $x_1$, the term $x_2^2$ is constant. The roles reverse when differentiating with respect to $x_2$.

At $𝕩=(3,2)^{\mathsf T}$, the two rates are 6 and 4.

### 3 · The gradient is a vector

For a differentiable scalar function of $p$ real coordinates, the **gradient** is the column of partial derivatives in input order:

$$
\nabla f(𝕩)=\begin{bmatrix}
\partial f/\partial x_1\\
\vdots\\
\partial f/\partial x_p
\end{bmatrix}.
$$

The symbol $\nabla$ is read “nabla.” For our two-coordinate function,

$$
\nabla f(𝕩)=\begin{bmatrix}2x_1\\2x_2\end{bmatrix}=2𝕩,
\qquad \nabla f(3,2)=\begin{bmatrix}6\\4\end{bmatrix}.
$$

The function value $f(3,2)=13$ is a scalar; the gradient has one component per input coordinate.

### 4 · Two basic vector differentiation rules

Let $𝕩,𝕨\in\mathbb R^p$, where $𝕨$ and the scalar $b$ are fixed. The notation $\nabla_{𝕩}$ states that we differentiate with respect to $𝕩$.

| Scalar expression | Gradient with respect to $𝕩$ |
|:---|:---|
| $𝕨^{\mathsf T}𝕩+b=\sum_{i=1}^{p}w_ix_i+b$ | $𝕨$ |
| $𝕩^{\mathsf T}𝕩=\sum_{i=1}^{p}x_i^2$ | $2𝕩$ |

In the first row, differentiating with respect to $x_i$ gives $w_i$. In the second, it gives $2x_i$. The second expression is the squared Euclidean norm $\lVert𝕩\rVert_2^2$.

For a worked combination, let $g(𝕩)=f(𝕩)+2x_1+x_2$. Then

$$
\nabla_{𝕩}g(𝕩)=\begin{bmatrix}2x_1+2\\2x_2+1\end{bmatrix},
\qquad \nabla_{𝕩}g(3,2)=\begin{bmatrix}8\\5\end{bmatrix}.
$$

The same sum rule applies: differentiate each term and collect matching components.
