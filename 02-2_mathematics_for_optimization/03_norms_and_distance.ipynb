## 02-2 · Part 3: Norms and Distance

**A cooling change needs a size rule before it can be called small or large.**

Part 2 compared the direction of $𝕕=(0.5,-0.5)^{\mathsf T}$. We retain that early-and-late cooling displacement and now measure its size and the distance between decisions.

<style>
.katex .textbb { font-style: italic; }
</style>

### 1 · Three ways to measure the same change

A norm assigns a nonnegative size to a vector, with size zero only for the zero vector. For $𝕕=(d_1,d_2)^{\mathsf T}=(0.5,-0.5)^{\mathsf T}$,

| Norm | Definition | Worked value |
|:---|:---|:---|
| Total absolute change | $\lVert 𝕕\rVert_1=\lvert d_1\rvert+\lvert d_2\rvert$ | $0.5+0.5=1$ |
| Euclidean length | $\lVert 𝕕\rVert_2=\sqrt{d_1^2+d_2^2}$ | $\sqrt{0.5}\approx0.707$ |
| Largest absolute component | $\lVert 𝕕\rVert_\infty=\max(\lvert d_1\rvert,\lvert d_2\rvert)$ | $0.5$ |

Absolute value $\lvert q\rvert$ is the nonnegative magnitude of a scalar $q$. The signed sum $0.5+(-0.5)=0$ would hide the two changes, so it cannot measure this displacement's size.

For a scalar $\alpha$ and another vector $𝕖$, norms satisfy

$$
\lVert\alpha 𝕕\rVert=\lvert\alpha\rvert\lVert 𝕕\rVert,
\qquad \lVert 𝕕+𝕖\rVert\le\lVert 𝕕\rVert+\lVert 𝕖\rVert.
$$

The first rule scales size by the multiplier's magnitude. The second is the triangle inequality: the combined change is no larger than the sum of the two change sizes.

### 2 · Distance depends on the size rule

The distance between two vectors is the norm of their difference. For the starting decision $𝕦=(3,2)^{\mathsf T}$ and the new decision $𝕧=(3.5,1.5)^{\mathsf T}$, the difference is $𝕧-𝕦=𝕕$. These coordinates set early cooling for $t=0,\ldots,5$ and late cooling for $t=6,\ldots,11$ in the 12-step schedule.

Thus the same two decisions are 1 unit apart in the total-absolute-change rule, about 0.707 in Euclidean length, and 0.5 in the largest-component rule.

Compare $𝕕$ with another displacement $𝕖=(0.75,0)^{\mathsf T}$:

| Change | $\lVert\cdot\rVert_1$ | $\lVert\cdot\rVert_2$ | $\lVert\cdot\rVert_\infty$ |
|:---|:---|:---|:---|
| $𝕕$ | 1 | 0.707, approximately | 0.5 |
| $𝕖$ | 0.75 | 0.75 | 0.75 |

The total absolute change is smaller for $𝕖$, but Euclidean length is smaller for $𝕕$. The coordinates stay fixed; only the size rule changes. Both coordinates here use the same cooling unit. Combining different physical units requires an explicit scaling rule.

### 3 · A unit direction

Separate direction from Euclidean length by dividing a nonzero vector by its length. For the same $𝕕$, define the unit vector $\hat{\vphantom{d}𝕕}$:

$$
\hat{\vphantom{d}𝕕}=\frac{𝕕}{\lVert 𝕕\rVert_2}
=\begin{bmatrix}1/\sqrt2\\-1/\sqrt2\end{bmatrix},
\qquad \lVert\hat{\vphantom{d}𝕕}\rVert_2=\sqrt{\tfrac12+\tfrac12}=1.
$$

The identity $𝕕=\lVert 𝕕\rVert_2\hat{\vphantom{d}𝕕}$ reconstructs the original change from its length and direction. For a nonnegative scalar $\ell$, the displacement $\ell\hat{\vphantom{d}𝕕}$ has length $\ell$. The zero vector cannot be normalized because its length is zero.
