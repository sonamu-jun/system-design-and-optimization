## 02-2 · Part 3: Norms and Distance

**A norm measures vector size; the norm of a difference measures distance.**

Part 2 used inner products to compare directions. We now measure the size of the displacement $𝕕=(1,-1)^{\mathsf T}$.

### 1 · Three common norms

A **norm** assigns a nonnegative size to a vector, with size zero only for the zero vector. For $𝕕=(d_1,d_2)^{\mathsf T}=(1,-1)^{\mathsf T}$,

| Norm | Definition | Value |
|:---|:---|:---|
| 1-norm | $\lVert𝕕\rVert_1=\lvert d_1\rvert+\lvert d_2\rvert$ | $2$ |
| 2-norm: Euclidean length | $\lVert𝕕\rVert_2=\sqrt{d_1^2+d_2^2}$ | $\sqrt2$ |
| Infinity norm | $\lVert𝕕\rVert_\infty=\max(\lvert d_1\rvert,\lvert d_2\rvert)$ | $1$ |

Absolute value removes a scalar's sign: $\lvert-1\rvert=1$. The signed sum $1+(-1)=0$ cannot measure this vector's size.

Norms also satisfy the following rules for a scalar $\alpha$ and vectors $𝕕,𝕖$ of the same dimension:

$$
\lVert\alpha𝕕\rVert=\lvert\alpha\rvert\lVert𝕕\rVert,
$$

$$
\lVert𝕕+𝕖\rVert\le\lVert𝕕\rVert+\lVert𝕖\rVert.
$$

The second rule is the triangle inequality: the size of a sum cannot exceed the sum of the sizes.

### 2 · Distance

For $𝕩=(3,2)^{\mathsf T}$ and $𝕪=(4,1)^{\mathsf T}$, their difference is $𝕪-𝕩=𝕕$. The Euclidean distance is

$$
\lVert𝕪-𝕩\rVert_2=\sqrt{(4-3)^2+(1-2)^2}=\sqrt2.
$$

The same points have distance 2 under the 1-norm and 1 under the infinity norm. The points stay fixed; the size rule changes.

### 3 · Unit vectors

A **unit vector** has Euclidean length 1. Divide a nonzero vector by its length to keep its direction and set its length to 1:

$$
𝕖=\frac{𝕕}{\lVert𝕕\rVert_2}
=\begin{bmatrix}1/\sqrt2\\-1/\sqrt2\end{bmatrix}.
$$

Indeed, $\lVert𝕖\rVert_2=\sqrt{1/2+1/2}=1$. The zero vector cannot be normalized because division by zero is undefined.
